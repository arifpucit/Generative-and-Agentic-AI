## MCP Servers: Local vs Remote Availability

**Legend**
- 🌐 **Remote**: the vendor hosts the server; you just add the URL to your MCP client (usually signing in with OAuth).
- 💻 **Local**: you run the server yourself, either through a stdio command (`npx`, `uvx`, `docker`) or as a local HTTP server.
- ❌: no official option of that type.

| Server | 🌐 Remote (vendor-hosted) | 💻 Local | Docs |
|---|---|---|---|
| **Notion** | ✅ `https://mcp.notion.com/mcp` (OAuth) | ✅ `npx -y @notionhq/notion-mcp-server` (integration token). Notion actively supports only the remote server and may retire this one. | [Notion MCP docs](https://developers.notion.com/docs/mcp) · [Local repo](https://github.com/makenotion/notion-mcp-server) |
| **Jira** (Atlassian Rovo) | ✅ `https://mcp.atlassian.com/v2/mcp` (OAuth 2.1). The old `/v1/sse` endpoint is no longer supported, and on 1 March 2027 existing v1 connections switch to the v2 tools. | ❌ No official local server. Stdio-only clients can bridge to the remote one with `npx -y mcp-remote <url>`. | [Getting started](https://support.atlassian.com/atlassian-rovo-mcp-server/docs/getting-started-with-the-atlassian-remote-mcp-server/) · [GitHub](https://github.com/atlassian/atlassian-mcp-server) |
| **Linear** | ✅ `https://mcp.linear.app/mcp` (OAuth 2.1); read-only at `/mcp/readonly` | ❌ No official local server; use the `mcp-remote` bridge if needed. | [Linear MCP docs](https://linear.app/docs/mcp) |

Not available for GitHub Enterprise Server. | ✅ `docker run -i --rm -e GITHUB_PERSONAL_ACCESS_TOKEN ghcr.io/github/github-mcp-server` | [GitHub Docs](https://docs.github.com/en/copilot/how-tos/context/model-context-protocol/using-the-github-mcp-server) · [Repo](https://github.com/github/github-mcp-server) |

| **Stripe** | ✅ `https://mcp.stripe.com` (OAuth or restricted API key) | ✅ `npx -y @stripe/mcp --api-key=YOUR_KEY` | [Stripe MCP docs](https://docs.stripe.com/mcp) |
| **PayPal** | ✅ `https://mcp.paypal.com/mcp`; sandbox at `https://mcp.sandbox.paypal.com/mcp` | ✅ `npx -y @paypal/mcp --tools=all` (access token + `PAYPAL_ENVIRONMENT`) | [mcp.paypal.com](https://mcp.paypal.com) · [Agent Toolkit](https://github.com/paypal/agent-toolkit) |

# <span style='background:lightgreen'>4. **Claude Desktop**</span>

## a. Overview

> **Claude Desktop** is Anthropic's native application for **macOS**, **Windows** and **Linux (beta)** that unifies **Claude Chat**, **Claude Cowork** and **Claude Code** in a single window — giving you AI assistance both in conversation and as an autonomous agent that can read files, run code and act on your computer.

**Key highlights:**
- Runs Anthropic's Claude models (Opus, Sonnet, Haiku) by default. It can also be pointed at a **third-party inference gateway**; the easiest one for students is **Ollama**, which lets you run open-weight models (Qwen, GLM, Kimi, DeepSeek, …) either locally on your machine or on Ollama Cloud (see the **Bonus** section at the end).
- Connects to local and remote tools, files and apps via **Desktop Extensions** (local, `.mcpb`) and **Web Connectors** (remote, OAuth).
- Three ways of working, in three tabs on the left pane:

| Tab | What it is | Free tier | Paid (Pro / Max / Team / Enterprise) |
|---|---|---|---|
| **Chat** | Conversational assistant with artifacts, projects, web search, memory | ✅ | ✅ |
| **Cowork** | Agentic knowledge work on your files and apps, no terminal (on Pro/Max it is being merged into the Chat box — "Cowork is now just Claude") | ❌* | ✅ |
| **Code** | Claude Code inside the app: agentic coding over a repository | ❌* | ✅ |

\* With Anthropic's models. When Claude Desktop is routed through **Ollama** (Bonus section) inference does not go through Anthropic and you can skip the Anthropic sign-in, so Cowork and Code become usable without a paid Anthropic plan.

📖 Reference: [Install Claude Desktop](https://support.claude.com/en/articles/10065433-install-claude-desktop)

## b. Download, Install and Update

**System requirements**

| OS | Minimum version | Notes |
|---|---|---|

| Windows | 10 or later | Full feature set; keep the app updated for Cowork |

**macOS / Windows**
1. Go to [https://claude.ai/download](https://claude.ai/download) and download the installer for your OS.
2. Open the file and complete the installation.
3. Launch **Claude** from **Applications** (macOS) or the **Start Menu** (Windows).
4. The app updates itself automatically.

**Linux (Ubuntu / Debian) — recommended apt route**
```bash
# 1. Import Anthropic's signing key and add the repository (copy the exact commands
#    from the install article — verify the key fingerprint with `gpg --show-keys`)
# 2. Install
$ sudo apt update && sudo apt install claude-desktop
# 3. Launch from the applications menu, or:
$ claude-desktop
```
Alternatively download the `.deb` and run `sudo apt install ./claude-desktop_*.deb` (this does **not** enroll you in automatic updates). On Linux the app does not update itself — use `sudo apt update && sudo apt upgrade`.

**Sign in**
- On first launch, sign in with your **claude.ai** account (email or Google). Your chats, projects and settings sync across web, desktop and mobile. A Free account is enough to get started.

## d. Interface Walk-Through: The Left Pane

The left pane is your primary navigation hub. Explore each of the following:

### i. New Chat
- Click **New Chat** to begin a fresh conversation (independent of prior threads by default).
- **Practice:** Start a new chat → rename it → pin it → share it with the **Share** button (top-right, generates a link) → add it to a Project with **"Add to Project"**.
- **Charts and widgets:** Ask Claude to turn data, ideas or processes into a visualization — no external tool needed. Enable under **Settings → Capabilities → Visuals → Inline visualizations**.

### ii. [Artifacts](https://support.claude.com/en/articles/9487310-what-are-artifacts-and-how-do-i-use-them)
- Artifacts let Claude generate shareable apps, tools, visualizations, code and documents in a side panel, separate from the chat.
- A **live artifact** connects to external data and refreshes every time it is opened — useful for dashboards and trackers you return to (built in Cowork).
- Available on all plans.
- **Practice:** **Settings → Capabilities → Enable Artifacts and AI-powered artifacts**, then ask Claude to *"create an interactive bar chart of Pakistan's population by province"*.

### iii. [Projects](https://support.claude.com/en/articles/9517075-what-are-projects)
- A Project is a self contained workspace, where you can give your own system prompt (instructions) and context by uploading files. It maintains a separate memory for the project, and within a project you can have multiple chat sessions.
- **Practice:** Create a Project called *"AI Research"* → upload a PDF paper → set a custom system prompt → start a chat inside it and observe that context persists across sessions.

### iv. [Claude Code](https://support.claude.com/en/articles/11145838-use-claude-code-with-your-pro-or-max-plan)
- Anthropic's agentic coding tool, integrated into the desktop app (Pro/Max/Team/Enterprise, or any plan via Ollama). It reads, edits and runs code across a whole repository with full transparency and permission prompts.
- **Practice (Pro/Max or Ollama):** Click the **Code** tab → select a project folder → ask *"read my project folder and explain its structure"*.

### v. [Claude Cowork](https://support.claude.com/en/articles/13345190-get-started-with-claude-cowork)
- Cowork brings Claude Code's agentic architecture to a **visual, no-terminal interface** for knowledge work: documents, research, spreadsheets, presentations, file management. Claude plans the task, breaks it into sub-tasks (with sub-agents) and executes them in an isolated virtual machine.
- Requires a paid plan (or Ollama). Cowork also exists on claude.ai, mobile and the Chrome side panel; on Pro/Max the separate Cowork box is being merged into Chat ("Cowork is now just Claude"), so your interface may differ slightly.
- Cloud sessions keep running if you close the app; only tasks that use **local files, the built-in browser or your computer** need the desktop app to stay open.
- **Practice:** Click **Cowork** → connect a folder → *"Summarize all PDF files in this folder into a single Word document"* → review the plan → let it run → inspect the output.

### vi. Pinned & Recent Chats
- Previously active or pinned chats appear in the **Pinned** and **Recent** sections.

### vii. Avatar (Settings)
- Click your avatar at the bottom of the left pane to open **Settings**:
    - **General:** name/preferences, notifications, appearance, voice, **Enable computer use** (Pro/Max).
    - **Account:** account details, sessions, log out / delete account.

# <span style='background :lightgreen' >7. Connecting Claude Desktop with <b> Local MCP Servers</b></span>
### i. What is a Local MCP Server?
- A program that **Claude Desktop starts** as a child process when it launches, and stops when it quits. The two talk over **stdio** (standard input/output) — there is no URL and no network port.
- It runs with **your user account's permissions**: it can do whatever you can do on the machine. Only install servers from sources you trust, and limit each one to the folders or commands it actually needs.
- It is **machine-scoped**: it works only in **Claude Desktop** on the computer where it is installed — not on claude.ai, Claude mobile, cloud Cowork sessions, or your other machines.
- Every tool call needs your **approval**: Claude shows the tool's name and its arguments and waits for you to allow or deny it. Read the request before you click.

> **Privacy note:** A local server needs no cloud account, which makes it a good fit for private folders, local databases and system resources. However, whatever a tool **returns** (for example, the contents of a file Claude reads) becomes part of the conversation and is sent to the model. Only give a server access to data you are willing to share with Claude.

### ii. The Three Techniques to Connect to a Local MCP Server in Claude Desktop

| § | Technique | Example server | Where the server's code comes from | How Claude Desktop learns to start it | You install |
|---|---|---|---|---|---|
| **7a** | **Desktop Extension (`.mcpb`)** | Filesystem | A ready-made bundle from the Extensions directory (or a `.mcpb` file) | The bundle's own `manifest.json` — you only fill in a settings form | Nothing (Node.js is built into Claude Desktop) |
| **7b** | **Config file + `uvx`** | Git | A package that `uvx` downloads from PyPI | An entry **you write** in `claude_desktop_config.json` | Git, uv |
| **7c** | **Clone from source + config file** | Desktop Commander (bash) | Source code you `git clone` and build on your machine | An entry **you write** in `claude_desktop_config.json`, pointing at your build | Git, Node.js |

Each technique gives you more control than the one before, and asks for more understanding: 7a is how most users install servers, 7b is how you add any published server, and 7c is how you will run the servers you write yourself later in this course.

### iii. Anatomy of a Config-File Entry (used in 7b and 7c)
```json
{
  "mcpServers": {
    "<server-name>": {
      "command": "<absolute path of the program to run>",
      "args": ["<argument 1>", "<argument 2>"],
      "env": { "<VARIABLE>": "<value>" }
    }
  }
}
```

| Key | Meaning |
|---|---|
| `mcpServers` | One object that holds **all** your local servers. Add each new server as another entry inside it — never create a second `mcpServers`. |
| `<server-name>` | A label you choose, e.g. `git`. It is shown under **Connectors** and names the server's log file (`mcp-server-git.log`). |
| `command` | The program Claude Desktop launches, given as an **absolute path** (7b explains why). |
| `args` | The command-line arguments passed to that program. |
| `env` | *Optional.* Environment variables for this server only. Values are stored as **plain text** — unlike extension settings, which are encrypted. |

Claude Desktop reads this file **only at startup**, and a single JSON error (e.g. a missing or trailing comma) prevents **every** server in it from loading.

## (ii) What is a `.mcpb` File?
- An **MCP Bundle** (`.mcpb`) is a zip archive containing a local MCP server **plus** a `manifest.json` that describes it. The format was previously called **Desktop Extensions** with the `.dxt` extension.

That is why this technique needs no config-file entry: the manifest replaces it. Settings marked `"sensitive": true` (like API keys) are encrypted in your OS's secure storage (Keychain on macOS, Credential Manager on Windows).
- Bundles can contain **Node.js** servers (recommended, because Node.js ships inside Claude Desktop — you install nothing), **Python** servers (including `uv`-managed ones), or **compiled binaries**.

**Route 2 — From a `.mcpb` file you already have**
Use this route for a bundle that is not in the directory, e.g. one downloaded from a developer's GitHub *Releases* page or one you built yourself.
1. Open Claude Desktop → **Settings → Extensions → Advanced settings**.
2. In the **Extension Developer** section, click **"Install Extension…"** and select the `.mcpb` file.

3. Complete any other required fields, save, and make sure the extension is **enabled**.
4. **Verify:** click the **`+`** button in the chat input box → **Connectors**. The `filesystem` tools should be listed. If they are not, fully quit and relaunch Claude Desktop.

## (vi) Add the Server Entry
- Claude Desktop reads /Users/arif/Library/Application Support/Claude/claude_desktop_config.json when it launches.
- For JSON entry of an `mcpServers`, it spawns the command as a child process and talks to it over stdio.
- `uvx` (mention full path) fetches the mcp-server-git package from PyPI into a temporary, cached virtual environment under uv’s cache, typically /Users/arif/.cache/uv/
```json
{
  "mcpServers": {
    "git": {
      "command": "/Users/arif/.local/bin/uvx",
      "args": ["mcp-server-git", "--repository", "/Users/student/dev/project"]
    }
  }
}
```
- `command` is the program to launch: **`uvx`**.
- `args` is the **package name** (`mcp-server-git`) followed by the server's own options.
- `--repository` locks the server to that one repository (any path inside it is accepted; the server uses the repository root). Without it, Claude passes a repository path in each tool call, so any repo you name could be accessed.
- On Windows, write paths with escaped backslashes, e.g. `"C:\\Users\\student\\dev\\project"`.

On launch, Claude Desktop runs your `command` with its `args`; `uvx` finds the package in its cache from step (iii) and starts the server.

## (i) About Desktop Commander — Tools and Risks
**Desktop Commander is open source (MIT) and written in TypeScript/Node.js**.

**Tools** (grouped):
- **Terminal:** `start_process`, `interact_with_process`, `read_process_output`, `list_sessions`, `list_processes`, `force_terminate`, `kill_process`
- **Files:** `read_file`, `read_multiple_files`, `write_file`, `write_pdf`, `create_directory`, `list_directory`, `move_file`, `get_file_info`, `start_search`, `get_more_search_results`, `stop_search`, `list_searches`, `edit_block`
- **Configuration:** `get_config`, `set_config_value`
- **Usage & feedback:** `get_usage_stats`, `get_recent_tool_calls`, `give_feedback_to_desktop_commander`

**Risks you must understand before installing it:**
- It runs commands in a **real shell**. Its own documentation calls its guardrails **"not a sandbox"**: command blocking and directory limits can be bypassed (e.g. via symlinks, command substitution or running code).
- It uses a **blocklist** (`blockedCommands`): every command is allowed **except** the listed ones. The default list blocks commands such as `sudo`, `su`, `dd`, `mkfs`, `mount`, `shutdown`, `reboot` and `passwd` — but **not** `rm`.
- `allowedDirectories` limits the **file tools only**, not terminal commands. Its default is empty (`[]`), which means **the whole filesystem**.
- **Telemetry is on by default** (pseudonymous usage data; the project says it excludes file contents, paths and command arguments).
- Its own tool-call log (`~/.claude-server-commander/claude_tool_call.log`) stores arguments **unredacted**.

- For stronger isolation the project also offers a **Docker** install, which keeps the shell inside a container.

We deal with these in step (viii), **before** using the server.

- `src/` — the **TypeScript source code**, which you can read and modify.
- `node_modules/` — the dependencies `npm install` downloaded.
- `dist/` — the **compiled JavaScript** that `npm install` built from `src/`. **`dist/index.js` is the file Claude Desktop will run.**

## (v) Find the Absolute Paths
This time Claude Desktop needs **two** absolute paths: the `node` runtime and your built `dist/index.js` (run inside the clone folder):
```bash
$ which node                  # e.g. /usr/local/bin/node   (Windows: where node)
$ echo "$(pwd)/dist/index.js" # e.g. /Users/student/dev/DesktopCommanderMCP/dist/index.js
```
> If you installed Node.js with **nvm**, the absolute path is essential (e.g. `/Users/student/.nvm/versions/node/v22.12.0/bin/node`), because Claude Desktop does not see nvm's `PATH`.

## (vi) Add the Server Entry
- Claude Desktop reads /Users/arif/Library/Application Support/Claude/claude_desktop_config.json when it launches.
- For JSON entry of an `mcpServers`, it spawns the command as a child process and talks to it over stdio.
- `uvx` (mention full path) fetches the mcp-server-git package from PyPI into a temporary, cached virtual environment under uv’s cache, typically /Users/arif/.cache/uv/
Open the config file as in section 7b (v) and add a `desktop-commander` entry **next to** the existing `git` entry, inside the same `mcpServers` object:
```json
{
  "mcpServers": {
    "git": {
      "command": "/Users/arif/.local/bin/uvx",
      "args": ["mcp-server-git", "--repository", "/Users/student/dev/project"]
    },
    "desktop-commander": {
       "command": "/Users/arif/.nvm/versions/node/v22.22.3/bin/node",
       "args": ["/Users/arif/dev/DesktopCommanderMCP/dist/index.js"]
    }
  }
}
```
Compare the two entries: `git` runs a **package** through `uvx`; `desktop-commander` runs **your own build** through `node`. Note the comma between the two entries.

Save the file.

## (vii) Fully Quit, Relaunch and Verify
Quit and relaunch Claude Desktop as in section 7b (vii), then click **`+`** → **Connectors** and check that the `desktop-commander` tools are listed. If they are not, read `mcp-server-desktop-commander.log` in the logs folder (section 7b (ix)).

## (viii) Lock It Down — in a Separate Chat
Desktop Commander keeps its own settings in a **second** config file, `~/.claude-server-commander/config.json`, which you change through chat. Do this in a **dedicated new chat** before any real work — the project advises this, because during normal work Claude may try to loosen settings such as `allowedDirectories` when it hits a restriction.
1. *"Show me the Desktop Commander configuration."* → `get_config`
2. *"Disable Desktop Commander telemetry."* → sets `telemetryEnabled` to `false`
3. *"Set allowedDirectories to ["/Users/student/dev"]."*
4. *"Add rm, curl, wget and ssh to blockedCommands, keeping the existing entries."*
5. *"Show me the configuration again."* → check that all four changes were saved.

## (ix) Use the Server
Start a **new** chat. Read **every** command in the approval prompt before allowing it.

> **Practice:**
> 1. *"Run `ls -la` in /Users/student/dev."*
> 2. *"Which versions of python3, node and git are installed?"*
> 3. *"Using grep, find every line containing TODO in the .py files in /Users/student/dev/project."*
> 4. *"Start a Python REPL, compute the sum of the squares of 1 to 100, then exit."* → `start_process` + `interact_with_process` driving an interactive program.
> 5. Test the blocklist: *"Run `rm /Users/student/dev/notes.md`."* → should be **rejected**, because you added `rm` to `blockedCommands`.

## (x) Update and Remove
- **Update:** `cd ~/dev/DesktopCommanderMCP && git pull && npm install` (pulls the new source and rebuilds `dist/`), then restart Claude Desktop.
- **Remove:** delete the `desktop-commander` entry from the config file, restart Claude Desktop, then delete the clone folder and `~/.claude-server-commander/`.

# Wrap-up: The Three Local Servers After Installation
You have now used all three techniques from section 7 (ii). Your `claude_desktop_config.json` contains the two servers you added by hand — the Filesystem server is **not** in it, because it is an extension managed by Claude Desktop:
```json
{
  "mcpServers": {
    "git": {
      "command": "/Users/student/.local/bin/uvx",
      "args": ["mcp-server-git", "--repository", "/Users/student/dev/project"]
    },
    "desktop-commander": {
      "command": "/usr/local/bin/node",
      "args": ["/Users/student/dev/DesktopCommanderMCP/dist/index.js"]
    }
  }
}
```

**Living with your servers:**

| | Filesystem (`.mcpb`) | Git (`uvx`) | Desktop Commander (clone) |
|---|---|---|---|
| **Where the code lives** | Managed by Claude Desktop | The uv cache | Your clone, e.g. `~/dev/DesktopCommanderMCP` |
| **How access is limited** | Allowed folders (strict) | One repository (`--repository`) | Command blocklist + `allowedDirectories` for file tools — **not a sandbox** |
| **Where its settings live** | Settings → Extensions | `args` in `claude_desktop_config.json` | `~/.claude-server-commander/config.json` (changed via chat) |
| **Logs** | `~/Library/Logs/Claude/` | `mcp-server-git.log` | `mcp-server-desktop-commander.log` |
| **Update** | Settings → Extensions | `uvx --refresh mcp-server-git --help` | `git pull && npm install` |
| **Remove** | Uninstall in Settings → Extensions | Delete the `git` entry | Delete the entry, the clone and `~/.claude-server-commander/` |

**Key takeaways**
- `claude_desktop_config.json` only tells Claude Desktop **how to start** a server; a server may keep its **own** settings elsewhere (as Desktop Commander does).
- Desktop Commander's file tools overlap with the Filesystem extension. Prefer the Filesystem extension for file work (it is strictly folder-scoped) and use Desktop Commander for running commands.
- Pick the simplest technique that works: an extension if one exists, `uvx`/`npx` for a published package, and a clone when you need to read, modify or build the code yourself.

# <span style='background:lightgreen'>Bonus: (Running Claude Desktop with Ollama)</span>

### Overview
- Claude Desktop has a built-in **third-party inference gateway** mode: all model calls go to a provider *you* configure instead of Anthropic's API, and the sign-in screen lets you **skip Anthropic authentication**. Enterprises use this with Bedrock / Vertex / Foundry; **Ollama** (v0.33.0; announced on the Ollama blog on 25 Aug 2026) packages the same mechanism into a single toggle for everyone.
- With Ollama you can use **any model in Ollama** — run **locally** on your laptop (free, private) or on **Ollama Cloud** (large models, free tier with one concurrent model; Pro \$20 / Max \$100 per month) — inside Chat, **Cowork** and **Code**. This makes the paid-only tabs usable for coursework without an Anthropic subscription.
- **History note:** the earlier `ollama launch claude-desktop` command (Ollama v0.15, May 2026) stopped working when Claude Desktop began rejecting non-Anthropic model IDs and was removed in v0.23.2. The v0.33 integration uses a local proxy that Claude Desktop treats as a standard gateway provider. Older tutorials describing `OLLAMA_API_KEY` + `ollama launch claude-desktop` are out of date.

### Prerequisites
- **Claude Desktop** installed and up to date.
- **Ollama ≥ v0.33.0** from [https://ollama.com/download](https://ollama.com/download) (`ollama --version` to check). **Currently macOS only** — Ollama has said Windows support is coming; Windows/Linux students can follow along in class or use a free Anthropic account for the Chat exercises.
- **For local models:** pull at least one model beforehand, sized for your RAM/VRAM, e.g.
```bash
$ ollama pull qwen3.5          # check https://ollama.com/library for current names and sizes
```
- **For cloud models:** sign in to Ollama (**Ollama app → Sign in**) and enable cloud models. No API key or environment variable is needed any more.

### Step-by-Step Setup

**Step 1 — Turn on the integration**
1. Open the **Ollama** app and select **Apps** (or the **Claude** entry).
2. Turn **Claude** on. Ollama writes Claude Desktop's third-party-gateway configuration for you (you can see it afterwards under **Settings → Developer → Configure Third-Party Inference…**) and offers to install Claude Desktop if it is missing.
3. Restart Claude Desktop when prompted. On the sign-in screen choose to **continue with your inference-provider configuration** (skip the Anthropic sign-in).

**Step 2 — Map models**
- In **Ollama → Settings → Apps → Claude**, pick which Ollama model backs each Claude model slot (e.g. *Opus* → a large cloud model, *Sonnet* → a mid-size model, *Haiku* → a small local model). One Ollama model can serve several slots. Click **Restart Claude** to apply.
- In Claude Desktop the normal model picker now shows your mapped models; **Auto** mode keeps working.

**Step 3 — Use Chat, Cowork and Code**
- Open the **Cowork** or **Code** tab — they are no longer locked. Sub-agents inherit the selected model. Web search works through Ollama's web search.

Some open models worth trying (verify current names and sizes at [https://ollama.com/library](https://ollama.com/library)):

| Model | Provider | Notes |
|---|---|---|
| `qwen3.5` | Alibaba | Multilingual; small variants run locally on a laptop |
| `glm-5.1` | Zhipu AI | Strong agentic coding and reasoning (cloud) |
| `kimi-k2.6` | Moonshot AI | Strong reasoning and coding (cloud) |
| `minimax-m2.7` | MiniMax | Agentic coding and tool use (cloud) |
| `deepseek-v4` | DeepSeek | Reasoning / coding (cloud) |

### Switching back to Anthropic's models
- Toggle **Claude off** in the Ollama app (or quit Ollama) — Claude Desktop restores your previous configuration on restart. You can flip between the two at any time.
- From the terminal:
```bash
$ ollama launch claude-desktop --restore          # restore the standard Anthropic profile
$ ollama launch claude-desktop --restore --yes    # …and auto-approve the restart
```

### What works and what to watch for
- ✅ Chat, Cowork, Code, sub-agents, Auto mode, web search (via Ollama), local **and** cloud models.
- ⚠️ Platform: macOS only at the time of writing.
- ⚠️ Quality depends on the model: small local models are fine for quick questions; use a large cloud model for Cowork/Code tasks.
- ⚠️ While the toggle is on, your conversations go to Ollama (local or Ollama Cloud), **not** to Anthropic — Ollama states telemetry is off by default and has a zero-data-retention policy for its cloud. Check your institution's policy on work devices.
- ⚠️ Features that depend on your Anthropic account (sync to claude.ai/mobile, Projects in the cloud, Anthropic-hosted scheduled tasks, Dispatch) are not available in gateway mode; conversation history is stored on your machine.
