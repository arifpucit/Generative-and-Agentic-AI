| **d. No execution capability** | It generates text; it cannot run anything. | Compute the exact standard deviation of a dataset, create `abc.txt` on your desktop, run a Python script, or insert a record into a database. | Code execution, file-system and shell tools, ideally inside a sandbox |


# <span style='background :lightgreen' >2. From Simple LLM App -> LLM workflow -> Agent -> Agentic AI System</span>
## a. Simple LLM App

<h2 align="center"><div class="alert alert-success" style="margin: 20px">A <strong>Simple LLM Application</strong> makes a single LLM call per request. The developer fixes the flow in advance: the model takes the prompt and any supplied context (system instructions, conversation history, user-provided documents) and generates a text response in one pass. It does not call tools, take actions, or loop. Any reasoning happens inside that single generation.</div></h2>

<div style="margin: 20px 40px; color: #7e22ce; text-align: left;">
<strong>Common use cases:</strong>
<ol>
<li><strong>Grammar and tone checker:</strong> Paste a draft email or essay and get back a corrected, more professional version (the Grammarly-style rewrite).</li>
<li><strong>Plain chatbot, summarizer or translator:</strong> Ask a general question, summarize a long article, or translate a paragraph into Urdu, all answered in one call.</li>
</ol>
</div>

## b. LLM Workflow
<h2 align="center"><div class="alert alert-success" style="margin: 20px">An <strong>LLM Workflow</strong> is a system in which developer-written code orchestrates one or more LLM calls, and often tools or retrieval steps, along a predefined path. The steps and branching logic are decided at design time, so the LLM fills in each step but does not decide what the next step is. RAG pipelines and extract → validate → summarize pipelines are typical examples.</div></h2>

<div style="margin: 20px 40px; color: #7e22ce; text-align: left;">
<strong>Common use cases:</strong>
<ol>
<li><strong>Customer-support bot over company docs (RAG):</strong> Every question follows the same path: search the knowledge base → pass the top results to the LLM → return an answer with sources.</li>
<li><strong>Support-ticket router:</strong> One LLM call classifies each ticket as billing, technical, or general. The code then sends it to the matching prompt template, which drafts a reply. The LLM never chooses the route; the developer's if/else logic does.</li>
</ol>
</div>


## c. AI Agent
<h2 align="center"><div class="alert alert-success" style="margin: 20px">An <strong>Agent</strong> is an LLM-based system in which the model itself controls the flow of execution. In a loop, it reasons about a goal, chooses and calls tools, observes the results, and adapts its next steps until it reaches a stopping condition: the goal is met, a step or cost budget is exhausted, or control is handed to a human. Patterns such as ReAct (Reason + Act) implement this loop. The degree of autonomy varies with the guardrails, permissions, and human approval gates the developer puts in place.</div></h2>

<div style="margin: 20px 40px; color: #7e22ce; text-align: left;">
<strong>Common use cases:</strong>
<ol>
<li><strong>Coding assistant (Claude Code, GitHub Copilot agent mode):</strong> Given "fix this failing test," it reads files, edits code, runs the tests, studies the errors, and retries until they pass.</li>
<li><strong>Deep research or shopping agent:</strong> Given "find the best laptop under $1,000" or "compare the top vector databases," it searches, opens pages, filters results, and decides when it has enough information.</li>
</ol>
</div>

## d. Agentic AI System
<h2 align="center"><div class="alert alert-success" style="margin: 20px">An <strong>Agentic AI System</strong> is a coordinated network of multiple specialized AI agents that work together through an orchestration layer, sharing memory and communicating with each other to autonomously accomplish complex, long-running objectives with minimal human intervention.</div></h2>

<div style="margin: 20px 40px; color: #7e22ce; text-align: left;">
<strong>Common use cases:</strong>
<ol>
<li><strong>Voice-based customer support team (e.g., bank, airline, or telecom helpline):</strong> A customer calls and says, "I was charged twice for my plan." A greeter agent turns the speech into text, works out the intent, and routes the call. A verification agent confirms the caller's identity, and a billing agent checks the account and issues the refund. A voice agent speaks each reply naturally. All of them share the caller's history and the conversation so far, so the customer never has to repeat themselves. If the case is complex or the caller is upset, an escalation agent hands the call to a human with a short summary.</li>
<li><strong>AI research team (e.g., Claude's Research feature):</strong> A user asks, "Compare the battery strategies of the top 3 EV makers." A lead agent plans the work and splits it into separate questions. Several search agents work in parallel, one per company or source type, and save their findings to a shared notebook. A fact-checker agent verifies the claims and sources, and the lead agent merges everything into one well-organized report. If a section is weak or has gaps, the lead agent sends the agents back out for more.</li>
</ol>
</div>

> **A note on terms.** Anthropic's *Building effective agents* uses **"agentic systems"** as an umbrella term for both workflows and agents. In this course, an **agentic AI system** means several agents working together. The components in Section 3 apply to all of them: a multi-agent system is built from the same parts as a single agent.

## d. LLM app vs LLM Workflow vs AI agent

A simple way to feel the difference is to ask one question: **who decides the next step?**

- **LLM app:** *you* decide. One prompt, one LLM call, one response. To go further, you read the answer and prompt again, so **the human is the loop**.
- **LLM workflow:** *your code* decides. The app makes several LLM calls, and often tool or retrieval calls, but the developer wrote the steps and branches before it runs. Anthropic's *Building effective agents* (December 2024) calls workflows *"systems where LLMs and tools are orchestrated through predefined code paths."*
- **AI agent:** *the LLM* decides. In a loop, it reasons, calls a tool, reads the result and chooses what to do next, until the goal is met. The same article describes agents as systems where *"LLMs dynamically direct their own processes and tool usage"*, which in practice are *"LLMs using tools based on environmental feedback in a loop."*

All three can use the **same model**. An agent is **not a different kind of model**: it is the *same* LLM, wrapped in a control loop with tools, memory, and a stopping rule.

| Dimension | LLM app | LLM workflow | AI agent |
|---|---|---|---|
| **Shape of work** | One prompt → one output | A fixed pipeline: step 1 → step 2 → … | A loop: reason → act → observe → repeat |
| **LLM calls per request** | One | Several, known in advance | Many, one per pass through the loop; not known in advance |
| **Who decides the next step?** | The human | The developer's code (`if / else`, fixed loops) | The LLM itself, at run time |
| **Tool use** | None | Fixed tool or retrieval calls at set points | Chooses which tool, when, and with what inputs |
| **State / memory** | The conversation so far | The output of each step, passed to the next | A growing record of thoughts, tool calls and results |
| **When does it stop?** | After one generation | When the last step finishes | When the model has enough to answer, or a guardrail (step or cost limit) stops it |
| **Self-correction** | No: errors reach the user | Only through checks the developer wrote (gates, retries) | Yes: it sees a bad result and tries another way |
| **Cost and latency** | Lowest, predictable | Moderate, predictable | Highest, variable |
| **Testing and debugging** | Easy | Easy: the same path on every run | Harder: the path can change from run to run |
| **Example** | "Summarise this paragraph." | Classify a support ticket, then reply with the matching prompt | "Find the air-fare from LHE to KHI and total it for 3 people." |

## e. Do you actually need an agent?

Before learning *how* agents work, it is worth knowing *when* to build one. Autonomy is not free: every extra pass through the loop adds cost, latency, and a chance of going wrong. Good practice is to **start with the simplest design that works** and add autonomy only when the task demands it.

<div style="text-align:center;">
    <img src="../images/agent-new-02.png"
         style="max-width:1500px; width:100%; height:auto; display:inline-block;">
</div>


| Use… | When… | Example |
|---|---|---|
| **Generative AI app** | One call is enough | Translate some text OR summarise a document |
| **Workflow** | Steps are known and repeatable | Extract invoice fields → validate → save to database |
| **Agent** | Steps depend on what is discovered along the way | "Research these three suppliers and recommend one" |

> **A loop alone does not make an agent.** A workflow may also call an LLM inside a `for` loop (for example: write → evaluate → rewrite, up to 3 times). It is still a workflow, because **your code** decides what happens next. The question to ask is *who chooses the next step*: your code, or the LLM?

> **Rule of thumb.** If you can draw the flowchart of the task *before* it runs, build a workflow. If the flowchart depends on what the model finds, you need an agent.

# <span style='background :lightgreen' >3. Key Components of an Agentic AI System</span>

An agent is **not a new kind of model**. A useful way to see it is the formula

<h3 align="center">Agent = Model + Harness</h3>

- The **model** is the LLM. It reasons about the goal and decides the next step.
- The **harness** is every piece of code, configuration and execution logic that is **not** the model. It builds what the model sees, runs the tools the model asks for, and keeps the whole system safe. Claude Code, OpenAI Codex and Cursor are all harnesses wrapped around a model. *(The term was popularised in 2026 by LangChain's article "The Anatomy of an Agent Harness".)*

         style="max-width:1500px; width:100%; height:auto; display:inline-block;">
</div>

**How to read the figure:**
- **Model (top, blue):** the LLM sits **outside** the harness. On every pass through the loop, the harness sends it **context**, and the model sends back a decision: a final **answer** or a **tool call**.
- **Harness (large panel):** plain code, grouped by colour:
    - **Orchestration (purple):** the **agent loop** that calls the model, runs tools, adds the results and repeats; may spawn **subagents** for big tasks.
    - **Context (green):** what the model sees on each call: **instructions**, **context/memory**, and **skills**.
    - **Action (orange):** what the agent can do: **tools**, **MCP servers** and a **sandbox** for running code safely.
    - **Safety (red):** what keeps it in bounds: **guardrails**, **human-in-the-loop** approval and **observability**.
- **User (left):** gives the goal, receives the answer, and is asked for **approval** before risky actions (the dashed line from the safety band).
- **Outside world (right):** the systems the tools act on. Their **results** flow back (dashed arrow) into the context for the next pass.

## (i) The model (LLM): the reasoning and decision-making engine

Not every LLM can drive an agent. What matters is the model's **capability level**:

| Capability | What it adds | Example models |
|---|---|---|
| **Text generation only** | One prompt → one response. It cannot call tools. Fine for simple Q&A, writing and summarising. | GPT-2 and GPT-3 base models, Llama 1 and 2 base, Mistral 7B base, Phi-1 and Phi-2 |
| **Tool (function) calling** | Instead of plain text, the model can return a **structured tool call**: a tool name plus JSON arguments. Your code runs the tool and sends the result back. **This is what makes an agent possible.** | GPT-4o onward, including the GPT-5 and GPT-6 families; Claude 3 onward, including the Claude 5 family; Gemini 2.5 and 3.x; Llama 3.1+ and Llama 4; DeepSeek-V3 and V4; Qwen 3.x; Kimi K2.5 and K3; GLM-4.7 and 5.x; gpt-oss |
| **Reasoning (thinking)** | Spends extra tokens *thinking* before it answers. Better at maths, code and multi-step plans, at the cost of higher latency. | OpenAI o-series, GPT-5.x and GPT-6 (`reasoning.effort`); Claude 4.x and 5 (extended / adaptive thinking); Gemini 2.5 and 3.x (thinking levels); DeepSeek-R1 and V4; Qwen 3 (`enable_thinking`); Phi-4-reasoning; Mistral Small 4; gpt-oss |

# Orchestration: who runs the loop

## (ii) The agent loop (the heart of the harness)
- The agent loop is **ordinary code** that repeats a few steps: **build the context → call the model → if the model asked for a tool, run it and add the result to the context → repeat**. It stops when the model gives a **final answer**, or when a **guardrail** (such as a step limit) stops it.
- The model decides *what* to do next; the loop makes it *happen*. Its key responsibilities are:
    - Sending the goal and the current context to the model on every step
    - Running the tools the model asks for, and feeding the results back
    - Handling retries, errors and fallbacks, so one failing tool does not crash the agent
    - Keeping control flow across many steps, and keeping the context within budget
- **Example:** For a travel-booking task, the orchestration layer:
    - Sends the user goal to the LLM
    - Invokes the flight search tool
    - Filters results using constraints
    - Asks the LLM to evaluate options
    - Calls the booking tool only when conditions are satisfied

## (iii) Subagents
- **What they are:** when a task is too big, or too noisy, for one context window, the main agent (the **orchestrator**, or *coordinator*) splits it and **delegates** the pieces to **subagents**. A subagent is a helper agent with:
    - its **own fresh context window**: it does *not* see the main conversation. It starts from a short **task brief** that the main agent writes for it;
    - its **own instructions and tools**, for example read-only tools for a search task, and sometimes a smaller, cheaper model;
    - its **own agent loop**: it reads, searches and calls tools as many times as it needs.
- When it finishes, it returns **only its final result**, usually a short summary, to the agent that created it. All the files it read and the tool calls it made stay in *its* context and are thrown away.
- **Why it helps:**
    - **A clean main context:** the orchestrator receives a one-paragraph summary instead of thousands of tokens of search results, which keeps it focused and avoids *context rot* (Section (v)).
    - **Parallel work:** independent subagents can run **in parallel**.
    - **Specialists:** each subagent gets a prompt, tools and permissions suited to one kind of job, and nothing more.
- **The cost:** every subagent uses its own tokens, and a summary can drop details. The brief must be self-contained, because the subagent knows nothing except what the brief tells it.

    <img src="../images/built-in-subagents-claude.png"
         style="max-width:1000px; width:100%; height:auto; display:inline-block;">

>- In Claude Code, you can also write your **own subagents**, which is a Markdown file with a `name`, a `description`, the allowed `tools`, and a system prompt in the body. 

**Can subagents talk to each other?**
- **By default, no.** Subagents form a **hub and spoke** (star): each subagent reports **only to the agent that created it**, and only **once**, when it finishes. Two sibling subagents never see each other's work. If subagent B needs what subagent A found, the orchestrator must pass it on in B's brief. Only the orchestrator sees every result, so only it can merge them.
- **Why this design:** it is simple, cheap and predictable, and it keeps each context clean. The orchestrator stays in control of the whole task.
- **Newer exceptions in Claude Code (2026):**
    - **Named subagents:** a subagent that is given a name can be messaged with the `SendMessage` tool. The main agent can send follow-up work to a finished subagent, and named subagents that have this tool can message each other. This works only inside one session.
    - **Agent teams** (experimental, off by default; turn on with `CLAUDE_CODE_EXPERIMENTAL_AGENT_TEAMS=1`): a **team lead** session spawns **teammates**. Each teammate is a separate Claude Code session that **messages the other teammates directly** and claims work from a **shared task list**. You can also talk to any teammate yourself. Teammates cannot start teams of their own.
- A subagent can also start subagents of its own, up to three levels below the main conversation by default.

| | Subagents (default) | Agent teams |
|---|---|---|
| **Who talks to whom** | Each subagent returns its result to its creator only | Teammates message each other directly |
| **Coordination** | The orchestrator manages all the work | Messages plus a shared task list; the lead combines the results |
| **Token cost** | Lower: only summaries come back | Higher: each teammate is a full Claude session |
| **Best for** | Focused jobs where only the result matters | Work where agents must share findings and challenge each other |

- **Examples:** in Claude's Research feature, a lead agent sends several search subagents out in parallel and merges their findings into one report. Claude Code uses the Explore subagent to search a large codebase without filling the main conversation. In the OpenAI Agents SDK, the same ideas appear as *agents as tools*, where a subagent returns its result to the caller, and *handoffs*, where control passes to another agent (Lec-50).

## (iv) Instructions (the system prompt)
- The **system prompt** tells the model its **role, goal, rules, tone and output format**. It is sent at the start of **every** call.
- Coding agents also load **project memory files** automatically at the start of each session: `CLAUDE.md` in Claude Code, and `AGENTS.md` in OpenAI Codex, Cursor and many other tools. These hold the build commands, coding conventions and mistakes to avoid for that project.
- Good instructions are **specific and short**: every line is resent, and paid for, on every call.

## (v) Context and memory (what the model remembers)
- Memory enables an agentic AI system to retain context, learn from past interactions, and behave consistently over time. Without memory, every conversation with your agent starts from zero, like talking to someone with amnesia. With memory, your agent remembers previous conversations, user preferences, past decisions and historical context. *(You will see this in code in Examples 2 and 3.)* The types of memory an agent uses are:
- **Short-term memory** in LLM applications is the information it has about recent interactions, typically the ongoing conversations with the user or the behavior of the LLM. In LLM applications, this manifests as the conversation history that persists across API calls and is continuously provided to the model as context.
    - **Key Characteristics:**
        - *Session-Based Persistence:* Memory exists only during an active session and must be re-fed with each interaction
        - *Context Window Constraints:* Implemented as conversation history within the prompt, constrained by the model's token limit
        - *Dynamic Management:* Requires active strategies to manage limited space as conversations grow
    - **Basic Implementation Technique (Send Full Conversation History with each API Call):** Stores and transmits the complete history of user messages and assistant responses with each API call.
        - If using *`Chat Completion API`* you maintain a list of all messages and include them in every request (rapidly consumes token budget).
        - If using *`Responses API`* you can pass `previous_response_id` (chains to the previous response) or `conversation` (a durable conversation object created with the Conversations API), and OpenAI keeps the history on the server side. The earlier turns are still sent to the model, so they still count (and are billed) as input tokens.
    - **Common Techniques to Keep the History within the Model's Context Window:**
        - *Sliding Window:* Retains only the most recent N messages, creating a "moving window" through the conversation. It can be implemented using a fixed-size buffer (e.g., FIFO queue) that automatically discards the oldest message when a new one arrives. It maintains recent context and is efficient for ongoing conversations. However, earlier details and context are permanently lost; may lose critical information from the beginning of conversations.
        - *Token-Based Truncation:* Manages memory based on token consumption rather than message count, maintaining a "token budget." It can be implemented by tracking cumulative token count of messages; when approaching the limit, remove oldest messages until within budget. This technique also discards important early information and requires token counting logic.
        - *Context Switching:* Dynamically adjusts focus based on detected topic changes, prioritizing relevant context. It can be implemented using topic modeling or classification to detect conversation shifts. The programmer needs to maintain separate context buffers for different topics or reset context when topics change. For example: When conversation shifts from travel planning to cooking recipes, the system deprioritizes flight details and focuses on cooking recipes.
        - *Conversation Summarization:* Generates compressed summaries of older conversation segments, similar to meeting notes. It can be implemented by periodically invoking the LLM to create concise summaries of older exchanges; replace original messages with summaries to save tokens. It requires additional API calls to LLM for summary generation and the risk of losing granular details still exists.
- **Long-term memory** is critical for building autonomous and intelligent agentic AI systems. Unlike short-term memory (which exists only during a session), it has the following characteristics:
    - Persists across multiple sessions, conversations, and interactions, enabling agents to recall past interactions
    - Should store information such as the user's name, age, likes and dislikes
    - Should store decisions made in the past
    - Should evolve over time implicitly or explicitly.
- **Where knowledge can live: parametric memory, and the three types of long-term memory that a harness manages**
    - **Parametric Memory (Model Weights), not managed by the harness:** Stores world facts, skills, rules and reasoning patterns inside model weights during training (e.g., knowing Islamabad is the capital of Pakistan, or being able to write Python code). Useful for global knowledge, not for personalized or evolving state. Changing it requires touching the weights — continued pretraining, fine-tuning (full or LoRA/adapters), RLHF/DPO, or direct model editing (ROME/MEMIT) — so it cannot be updated at runtime, is opaque and un-auditable, and is impractical for per-user or fast-changing data.
    - **Semantic Memory:** Stores decontextualised facts — domain knowledge, documents, policies, manuals, and facts about the user (name, role, "prefers metric units", "allergic to peanuts"). Unlike episodic memory, it keeps the fact and discards when and how it was learned. Commonly implemented with RAG over a vector database, but equally valid as a knowledge graph, SQL table, or plain files.
    - **Episodic Memory:** Stores specific past events with their context — timestamps, entities involved, actions taken, outcomes and failures (e.g., "last Tuesday the student attempted matrix inversion and got the cofactor step wrong", or "the deploy failed because the config file was missing"). Enables case-based reasoning; in practice often surfaced as few-shot examples of past trajectories.
    - **Procedural Memory:** Stores *how* to do things — step-by-step workflows, plus rules, heuristics, policies and learned behavioural patterns. Typically encoded in system prompts, playbooks, runbooks, or code.
- **Context engineering.** Everything the model sees on a call (instructions, tool descriptions, conversation history, tool results, retrieved memories) shares one context window. Deciding what goes into that window, and what stays out, is called **context engineering**. More is not always better: as the input grows, models use it less reliably, even on simple tasks, an effect known as **context rot** (Chroma, 2025). This is why harnesses compact old messages, trim long tool results, load skills only when needed, and hand big subtasks to subagents. *(Examples 4 and 5 measure the context window growing.)*

## (vi) Skills (know-how, loaded only when needed)
- **What a skill is:** a **folder** that packages the **know-how for one kind of task**: a `SKILL.md` file with instructions, plus any scripts, reference documents and templates that the task needs. Anthropic compares writing a skill to *"putting together an onboarding guide for a new hire."* Like a standard operating procedure (SOP), it tells the agent:
    - the steps to go through
    - the format the result must follow (for example, how to build a report)
    - the scripts or tools to use, and how to check the result
- **What is inside the folder** (from the *Agent Skills* specification):

| Part | Required? | What it holds | How the agent uses it |
|---|---|---|---|
| `SKILL.md` | **Yes** | YAML frontmatter (`name`, `description`) and the Markdown instructions | **Reads** the instructions when the skill is activated |
| `scripts/` | No | Executable code: Python, Bash, JavaScript… | **Runs** it. Only the output enters the context, not the code |
| `references/` | No | Detailed docs: checklists, API guides, schemas | **Reads** a file only when a step needs it |
| `assets/` | No | Templates, images, fonts, sample data | **Uses** them to build the output |

```markdown
4. Run scripts/check.py to check the report, and fix anything it flags.
5. Cite sources as described in references/style.md.
```

**How an agent uses a skill**

A skill is only files on disk. Whatever runs the agent has to do three jobs, one for each level of progressive disclosure:
1. **Discover:** read only the `name` and `description` of every skill, and list them in the system prompt.
2. **Activate:** when a request matches a description, load the whole `SKILL.md` body into the context.
3. **Execute:** read the files and run the scripts that the instructions point to.

| Job | An agent written in plain Python (no framework) | Claude Code |
|---|---|---|
| **1. Discover** | Your code scans a `skills/` folder at startup, reads the frontmatter of each `SKILL.md`, and adds a list such as *"lab-report: Write a GenAI lab report… Use when…"* to the system prompt. | Done automatically for every skill in `~/.claude/skills/` and `.claude/skills/`. |
| **2. Activate** | You give the model a tool, such as `load_skill(name)`, that returns the `SKILL.md` body. The model calls it when a request matches a description. | Claude loads the skill on its own when a request matches its description, or you type `/lab-report`. |
| **3. Execute** | You give the model tools such as `read_file(path)` for `references/` and `assets/`, and `run_script(path)` for `scripts/`, ideally inside a sandbox (ix). | Claude uses its built-in tools to read files and run commands. |

>The agent loop does not change. A skill is just a short list in the system prompt plus a few ordinary tools, so it works with any model that supports tool calling.

# Action: what the agent can do

## (vii) Tools (the hands that let the agent act)
- A **tool** is a function the model can ask to use: search the web, query a database, run code, send an email. Tools let the model *do* things, not just *know* things (Section 1).
- **The model never runs a tool itself.** Instead of plain text, it replies with a **tool call**: a structured request that names the tool and gives its arguments. The harness runs the tool and sends back the result, which the model reads on its next pass. You will see this pattern, in plain text, in the ReAct example in Section 6.
- **How the model knows about a tool:** with each request, the harness sends a short **tool definition** for every tool: a `name`, a `description`, and the `parameters` as a JSON Schema. The model sees only this definition, never the function's code. A clear description is what makes it pick the right tool at the right time.

## (viii) MCP servers (one standard plug for tools and data)
- **The problem:** without a standard, every AI app writes its own integration for every service it wants to use (GitHub, Gmail, a database…). With M apps and N services, that is **M × N** integrations to build and maintain.
- **MCP (Model Context Protocol)** is an open standard, introduced by Anthropic in November 2024, for connecting agents to external tools and data through one uniform interface. Each service is wrapped **once** as an **MCP server**, and each app learns MCP **once**, so only **M + N** pieces are needed. The MCP documentation compares it to a **USB-C port for AI applications**.
- In December 2025, Anthropic donated MCP to the **Agentic AI Foundation**, a fund under the Linux Foundation co-founded with Block and OpenAI. MCP is supported by Claude, ChatGPT, Gemini, Microsoft Copilot, VS Code, Cursor and many other tools.
- **Safety:** an MCP server runs code and can reach your data. Connect only to servers you trust. The MCP specification also says that the host must get the user's consent before it invokes a tool.

         style="max-width:1200px; width:100%; height:auto; display:inline-block;">

- **Host:** the AI application you work in, such as Claude Code, ChatGPT, VS Code, or an agent you write yourself. It contains the LLM.
- **MCP client:** a connector inside the host. There is **one client for each server**.
- **MCP server:** a program that exposes one service, such as your files, GitHub or a database, and talks to that service for the agent.
- The client and server exchange **JSON-RPC** messages over one of two transports. With **stdio**, the host starts the server as a process on your own machine. With **Streamable HTTP**, the server runs elsewhere, at a URL. The numbered steps inside the host show a tool call:
    1. The client asks the server for its tools (`tools/list`).
    2. The LLM picks a tool.
    3. The client asks the server to run it (`tools/call`).
    4. The result goes back to the LLM.

>- *(MCP is covered in detail in Lec-44 to Lec-47.)*

## (ix) Sandbox (a safe place to run code)
- **Why it is needed:** when an agent runs code or shell commands, reads and writes files, or browses the web, it acts with **your** permissions. A mistake (an `rm -rf` in the wrong folder), or a **prompt-injection** attack hidden in a web page or a README, could then delete files, leak secrets or install malware.
- **What a sandbox is:** an **isolated environment** with a boundary around the code the agent runs. Inside it, the agent gets only what the task needs:
    - **files:** usually only the project folder is writable;
    - **network:** connections only to an allow-list of domains;
    - **credentials:** no secrets, or short-lived tokens limited to the task.

  Ideally the sandbox is created when the agent needs it and thrown away afterwards.
- **A sandbox and permission prompts work together:** a permission prompt (x) decides **whether** an action runs, and the sandbox limits **what the action can reach** once it runs. That is why an agent inside a sandbox can safely run many commands without asking you each time.

# Safety: what keeps it in bounds

## (x) Guardrails, human-in-the-loop and observability
- As agents become more autonomous, they gain the power to take **real actions** in the world, so **not every action should be automatic**. Imagine an agent that can send emails on your behalf, make purchases, delete files or post on social media.
- **Hooks/Guardrails** are automatic checks at fixed points around the agent. A check can be plain code, such as a regular expression or a list of allowed commands. It can also be a second, smaller LLM that acts as a classifier. Each check lets the step *pass*, *blocks* it, or hands it to a *human*. There are two kinds:
    - *Limits and permissions:* step, time and cost limits, and which tools, folders and websites the agent may use.
    - *Checks on content* at four checkpoints:
        - on the *input*, for example prompt injection or off-topic requests;
        - *before a tool runs*, for example `rm -rf` or a write outside the project;
        - *after a tool runs*, for example secrets in a result;
        - on the *output*, for example a wrong format or leaked data.
>- *Hooks* are Claude Code's way to add your own guardrails. A hook can be a script that runs every time its event fires. In the OpenAI Agents SDK, a guardrail that fails trips a tripwire and the SDK raises an exception and stops the run.
- **Human-in-the-loop (HITL):** for risky or irreversible actions, the agent **proposes** the action and **waits for a human to approve it** before executing. In Claude Code, this is the permission prompt you see before a command runs.
- **Observability:** keep a full record of every step, tool call, result, cost and delay (logs and traces). When something goes wrong, you can see exactly what the agent did, and you have a record for review and accountability.

# <span style='background :lightgreen' >4. Hands-On: Simple LLM Apps</span>

## <span style='color:#7e22ce'>Example 1: A single LLM call (no memory, no tools)</span>

- This is the app we have built since the first lectures: **prompt in, answer out**. The loop only lets you ask several questions in a row; **each question is one independent LLM call**. The model answers only from what it learned in training. It has no way to check a fact, and it gives no sign of which parts it is unsure about.
- **How the code works**
    - Each pass of the `while` loop sends a `messages` list that holds **only the current question**, so every call starts from nothing.
    - `display(Markdown(...))` shows the answer as formatted text (headings, tables, bold) instead of raw Markdown.
- The app has no real-time awareness. However, since Groq quietly adds the current date to the hidden system prompt of `gpt-oss` models, it may give you the current date correct. Ask the questions in following question:
```
Compare the University of the Punjab and the Lahore University of Management Sciences (LUMS): the year each was established, its age in 2026, its number of students, and who founded LUMS. Which one is older, and by how many years?
```

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    # ONE LLM call per question: prompt in, answer out. No tools, no memory.
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user", "content": user_input}],
    )
    display(Markdown(response.choices[0].message.content))
````

## <span style='color:#7e22ce'>Example 2: A chatbot without memory (each call sends only the latest message)</span>

- An LLM has **no memory of its own**. Every API call is independent: the model sees only the `messages` you send in *that* call. A chatbot seems to "remember" only because the app sends the earlier conversation back every time. That list of messages is what fills the model's **context window**, and it is the agent's **short-term memory**.
- This LLM app has no real-time awareness. However, since Groq quietly adds the current date to the hidden system prompt of gpt-oss models, it may give you the current date correct. Ask the questions in following sequence and observe the output:
```
You:  Hi, I'm Arif and I teach Data Science at the University of the Punjab.
Bot: Nice to meet you, Arif! I’d be happy to help with any Data Science topics or questions you have.
You:  What's my name, and what do I teach?
Bot: I’m not able to see that information—could you remind me of your name and what you teach?
You:  What is the date today?
Bot: Today is October 23, 2026.
You:  exit
```

````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

system = {"role": "system", "content": "You are a helpful assistant. Answer in one or two sentences."}

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    # NO memory: every call sends only the system prompt and the latest message
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[system, {"role": "user", "content": user_input}],
    )
    print("Bot:", response.choices[0].message.content)
````

## <span style='color:#7e22ce'>Example 3: A chatbot with memory (each call sends the whole conversation so far)</span>

- The same chatbot, with the same model and system prompt. The only change is in how the app handles `messages`: it now **keeps the conversation** and sends all of it on every call.
- **How the code works**
    - `messages = [system]` is created **once, before the loop**, so it survives from one turn to the next.
    - Each turn appends your message, sends the **whole list** to the model, and then appends the model's reply as an `assistant` message.
- This LLM app has no real-time awareness. However, since Groq quietly adds the current date to the hidden system prompt of gpt-oss models, it may give you the current date correct. Ask the questions in following sequence and observe the output:
```
You:  Hi, I'm Arif and I teach Data Science at the University of the Punjab.
Bot: Nice to meet you, Arif! I’d be happy to help with any Data Science topics or questions you have.
You:  What's my name, and what do I teach?
Bot: Your name is Arif, and you teach Data Science at the University of the Punjab.
You:  What is the date today?
Bot: Today is October 23, 2026.
You:  exit
```

````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

system = {"role": "system", "content": "You are a helpful assistant. Answer in one or two sentences."}
messages = [system]                                      # the conversation so far = the context window

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    messages.append({"role": "user", "content": user_input})
    response = client.chat.completions.create(model="openai/gpt-oss-120b", messages=messages)

    reply = response.choices[0].message.content
    messages.append({"role": "assistant", "content": reply})   # remember the reply too
    print("Bot:", reply)
````

## <span style='color:#7e22ce'>Example 4: Watching the context window grow (full history)</span>

- Sending the whole conversation works, but it has a cost: **every turn resends everything before it**, so the input grows with each turn.
- Following example code is Example 3's chatbot with one extra `print`: after each reply it shows how many messages were sent and how many **input tokens** they used, read from `response.usage.prompt_tokens`.
- **Questions to ask:**
```
You: Hi, I'm Arif and I teach Data Science at the University of the Punjab.
You: Suggest a good first Python library for my students.
You: And one for plotting?
You: Which of the two should they learn first?
You: Give one tip for teaching it.
You: What's my name, and where do I teach?
You: exit
```
- **What to notice:** As the messages list grows, the input tokens rise on every turn. The bot still knows your name at the end. In a very long chat, or in an agent that adds a tool result on every step, this list keeps growing. Each call gets slower and more expensive, and when the list reaches the model's context-window limit, the call fails.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

system = {"role": "system", "content": "You are a helpful assistant. Answer in one or two sentences."}
messages = [system]                                      # FULL HISTORY: every message is sent on every call

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    messages.append({"role": "user", "content": user_input})
    response = client.chat.completions.create(model="openai/gpt-oss-120b", messages=messages)

    reply = response.choices[0].message.content
    print("Bot:", reply)
    print(f"     [messages sent: {len(messages)} | input tokens: {response.usage.prompt_tokens}]")
    messages.append({"role": "assistant", "content": reply})
````

## <span style='color:#7e22ce'>Example 5: Keeping the context small with a sliding window (only the last 4 messages)</span>
- The **sliding window** is one of the techniques used to manage the short-term memory. It keep the whole conversation, but send the model only the most recent `N` messages.
- **Questions to ask:**
```
You: Hi, I'm Arif and I teach Data Science at the University of the Punjab.
You: Suggest a good first Python library for my students.
You: And one for plotting?
You: Which of the two should they learn first?
You: Give one tip for teaching it.
You: What's my name, and where do I teach?
You: exit
```
- **What to notice**
    - From the third turn on, the messages sent stay at 5 and the input tokens **stop growing**: they level off at about 200 to 230 instead of climbing towards 400.
    - The price is memory. By the last question, your first message has dropped out of the window, so the bot has **forgotten your name**.
    - This is the core trade-off of **context management**. Real harnesses use smarter techniques, such as summarising (compacting) old messages, trimming long tool results, and giving big subtasks to subagents with their own context windows. Even when everything fits, very long inputs make models less reliable (*context rot*), so a smaller, cleaner context is usually better.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

system = {"role": "system", "content": "You are a helpful assistant. Answer in one or two sentences."}
history = []                                             # the whole conversation, kept by our code
WINDOW = 4                                               # but only the last 4 messages are sent

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    history.append({"role": "user", "content": user_input})
    messages = [system] + history[-WINDOW:]              # SLIDING WINDOW: what the model will see
    response = client.chat.completions.create(model="openai/gpt-oss-120b", messages=messages)

    reply = response.choices[0].message.content
    print("Bot:", reply)
    print(f"     [messages sent: {len(messages)} | input tokens: {response.usage.prompt_tokens}]")
    history.append({"role": "assistant", "content": reply})
````

## <span style='color:#7e22ce'>Example 6: A workflow with three LLM calls in a fixed sequence (prompt chaining)</span>

- The following example uses **prompt chaining** pattern and makes **three** LLM calls, one after another. **Each call works on the output of the previous one**.
- The point to note that **our code** decides that there are exactly three steps, in this order. The LLM never decides what to do next.

```text
question ──► [LLM call 1: draft] ──► draft ──► [LLM call 2: review] ──► review ──► [LLM call 3: rewrite] ──► final
```
- **How the code works**
    - `ask_llm(prompt)` makes **one** LLM call and returns the text of the answer.
    - The loop calls it three times, and each prompt contains the output of the call before it. Each result is shown under its own heading: *Call 1: Draft*, *Call 2: Review* and *Call 3: Final*.
- **Questions to ask:** (Ask same question as in Example 1, so you can compare the answers):
1. *Compare the University of the Punjab and the Lahore University of Management Sciences (LUMS): the year each was established, its age in 2026, its number of students, and who founded LUMS. Which one is older, and by how many years?*
> **What to notice**
>- The final text is usually clearer than Example 1, and the review step often **spots** doubtful facts and marks them with *(?)*.
>- But it cannot **fix** them. No new information enters the chain: every call works only from what the model already "remembers". Three guesses in a row are still guesses.
>- **More LLM calls do not make an app an agent.** The steps are fixed in advance, so this is a *workflow*. To fetch facts, the model needs **tools**; to keep going until it has the facts, it needs a **loop**.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

def ask_llm(prompt):
    """Make ONE LLM call and return the text of the answer."""
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    # LLM call 1: write a first draft
    draft = ask_llm(user_input)
    display(Markdown(f"### Call 1: Draft\n{draft}"))

    # LLM call 2: review the draft (input = output of call 1)
    review = ask_llm(f"Review this answer. List any facts that may be wrong or doubtful, and any unclear wording.\n\n{draft}")
    display(Markdown(f"### Call 2: Review\n{review}"))

    # LLM call 3: rewrite the draft using the review (input = outputs of calls 1 and 2)
    final = ask_llm(f"Rewrite the answer using the review. Remove or mark (?) any doubtful fact. "
                    f"Return only the improved answer.\n\nANSWER:\n{draft}\n\nREVIEW:\n{review}")
    display(Markdown(f"### Call 3: Final\n{final}"))
````

## <span style='color:#7e22ce'>Example 7: A workflow with routing (the LLM classifies, your code chooses the route)</span>

- The following example uses **routing** pattern: A support-ticket router for an internet provider. For each ticket:

1. **LLM call 1** classifies the ticket as `billing`, `technical` or `general`.
2. **Our code** looks up the matching system prompt in the `ROUTES` dictionary. A label it does not recognise falls back to `general`.
3. **LLM call 2** writes the reply using that specialised prompt.

```text
ticket ──► [LLM call 1: classify] ──► label ──► our code picks a prompt ──► [LLM call 2: reply] ──► answer
```
- **Questions to ask** (type each one at the `Ticket:` prompt, then type `exit` to finish the cell):
    -  *I was charged twice for my monthly package this month.*
    -  *My Wi-Fi keeps disconnecting every 10 minutes since yesterday.*
    -  *Thank you, your technician was very helpful yesterday!*
- Each one is a customer's support ticket. The output shows the **route chosen by our code**, followed by the reply.
> **What to notice**
>- The three tickets should go to `billing`, `technical` and `general`, and each reply follows the rules of its own prompt.
>- The LLM did two small jobs, *classify* and *write*, but **the path was drawn by the developer**: classify first, then exactly one of three prompts. The LLM cannot add a step, skip one, or decide to look up the customer's account.
>- The `if label in ROUTES else "general"` line is a small **guardrail**: if the model answers with something unexpected, the code still takes a safe path.
>- Like Example 6, this is a **workflow**. It is predictable, cheap and easy to test, which is why it is the right design when the steps are known in advance.

````python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

# One specialised system prompt per route. The developer writes these at design time.
ROUTES = {
    "billing":   "You are a billing specialist at an internet provider. Apologise for the problem, explain that a "
                 "refund takes 5 to 7 working days, and keep the reply under 80 words.",
    "technical": "You are a technical support engineer at an internet provider. Give at most three clear "
                 "troubleshooting steps, and keep the reply under 80 words.",
    "general":   "You are a friendly customer-service agent at an internet provider. Answer briefly and politely, "
                 "in under 60 words.",
}

def ask_llm(system_prompt, user_message):
    """Make ONE LLM call with a system prompt and return the text of the answer."""
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "system", "content": system_prompt},
                  {"role": "user", "content": user_message}],
    )
    return response.choices[0].message.content.strip()

while True:
    user_input = input("Ticket: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    # LLM call 1: classify. The LLM only fills in a label.
    label = ask_llm("Classify the customer ticket as exactly one word: billing, technical or general. "
                    "Reply with that one word only.", user_input).lower()

    # YOUR CODE chooses the route. The LLM never decides what happens next.
    route = label if label in ROUTES else "general"

    # LLM call 2: write the reply with the specialised prompt
    reply = ask_llm(ROUTES[route], user_input)
    display(Markdown(f"**Route chosen by our code:** `{route}`\n\n{reply}\n\n---"))
````

# <span style='background :lightgreen' >6. Hands-On: The ReAct Loop</span>

<h2 align="center"><div class="alert alert-success" style="margin: 20px">In Examples 1 to 7, <b>you or your code</b> decided every step. An agent hands that decision to the <b>LLM</b>, inside a loop. The best-known pattern for that loop is <b>ReAct</b></div></h2>

## a. What is ReAct?
- **ReAct = Reasoning + Acting.** It is a framework in which an AI agent **interleaves thinking and tool use in a loop** to solve complex tasks. It combines two ideas:
    - **Reasoning:** the model thinks through what it knows and what it still needs.
    - **Acting:** the model calls a tool (search, calculator, code runner, database…) to get *real* information or to take an action.
- These two steps **alternate** until the agent has enough to give a final answer.
- **Paper:** [ReAct: Synergizing Reasoning and Acting in Language Models (2023)](https://arxiv.org/abs/2210.03629)

## b. The ReAct loop

<div style="text-align:center;">
    <img src="../images/agent-new-03.png"
         style="max-width:1300px; width:100%; height:auto; display:inline-block;">
</div>

**How to read the figure:**

1. **Input:** the user's question arrives together with the **system prompt** and the list of **actions** the model may take.
2. **Think:** the model writes out its reasoning: what it knows, and what is still missing. If it already has enough, it answers.
3. **Act:** the model asks for an action with a specific input, such as `lookup[LUMS]`. Our code runs it in the outside world.
4. **Observe:** the result (or an error message) is **appended to the context**, so the model "sees" it on the next pass.
5. **Repeat or stop:** the model thinks again. The loop ends when the model gives a **final answer**, or when a **step limit** stops it safely.

The table below sums up what happens at each step, and why it matters.

| Step | What happens | Why it matters |
|---|---|---|
| **Think** | The LLM writes out its reasoning: *"I need to find X. I'll use the search tool."* | This reasoning is what separates ReAct from blind tool-calling: the model explains **why** it is acting, and can plan ahead. |
| **Act** | The model calls a tool with specific inputs, such as `search("PIA ticket Lahore to Karachi")` or `calculator("42 * 7")`. | The model reaches beyond its training data: live facts, exact maths, real systems. |
| **Observe** | The tool returns a result, which is **appended to the context**. The model now "sees" real-world output. | The next thought is grounded in facts instead of guesses. Errors are observed too, so the model can recover. |
| **Loop** | The model thinks again based on the new observation. It may need another tool, or it may be done. | The agent can self-correct and handle multi-step problems one step at a time. |
| **Final answer** | The model stops calling tools and answers the user. | The loop ends when the model has enough, or when a guardrail stops it. |

> **Key insight of ReAct.** Interleaving reasoning with actions makes agents far more reliable. The thinking step lets the model **self-correct**, **plan ahead**, and **avoid hallucinating facts it could simply look up**. As a bonus, the written reasoning makes the agent's behaviour **readable and debuggable**.

## c. <span style='color:#7e22ce'>Example 8: A ReAct agent from scratch</span>

The whole agent has **three parts**, and each one maps to a component from Section 3:

1. **An action** (the *tools* component): `lookup(name)` looks up a fact about a university in a tiny "database", a Python dictionary. The founding years match Wikipedia: the University of the Punjab in 1882, LUMS in 1985.
2. **Instructions:** the system prompt teaches the model the ReAct format, with one worked example, as in the original paper: write a `Thought:` and an `Action: lookup[...]`, then stop and wait for the `Observation:`. When it knows the answer, write `Final Answer:`.
3. **The ReAct loop** (the *agent loop* component): our code sends the conversation to the model, prints its Thought and Action, runs the lookup, and sends the result back as `Observation: ...`. If the model's reply contains no `lookup[...]`, the model has finished.

The model here writes plain text, and our code finds the action with a **regular expression**. This is exactly how the 2022 ReAct paper worked. Modern agents use **native tool calling** instead (Lec-40 and Lec-41), but the loop is the same.

> **Why this example uses `qwen/qwen3.8-27b` instead of `gpt-oss-120b`.** `gpt-oss` is a *reasoning* model: it does its thinking in a hidden channel. Asked to write `Thought:` and `Action:` lines, it often keeps them in that hidden channel and then invents the `Observation:` itself. Qwen 3.8, with its thinking turned off (`reasoning_effort="none"`), writes every Thought in the visible text, so you can watch the whole ReAct loop. It is also free on Groq.

**Questions to ask** (type each one at the `You:` prompt, then type `exit` to finish the cell):
1. *Which is older, the University of the Punjab or LUMS, and by how many years?*
2. *Who founded LUMS?*
3. *What is 2 + 2?*

**What to notice in the trace**
- Each `Thought:` + `Action:` pair is **one pass through the loop**, which means **one LLM call**. The `Observation:` line that follows was produced by **our code**, not by the model.
- **Question 1** needs two passes with a lookup, then a third that gives the final answer: *look up PU → look up LUMS → 1985 − 1882 = 103 years*. Nobody wrote this plan; the model chose each action after reading the previous observation. Compare the answer with Examples 1 and 6, which guessed the facts from memory.
- **Question 2** needs one lookup. **Question 3** needs none, so the loop ends on the first pass. Sometimes the model writes something other than `lookup[...]` after `Action:`, such as `Action: Final Answer: 4`; since it is not a lookup, our loop still treats the reply as the end.
- Three small lines keep the agent in bounds: `range(1, 6)` stops a loop that never finishes, `stop=["Observation:"]` stops the model from making up its own observations, and `lookup()` returns `"No information found."` instead of crashing when a name is unknown.
- `messages` grows on every pass: instructions, question, Thought + Action, Observation, Thought + Action, Observation, … This is the context-window growth you measured in Examples 4 and 5.

````python
import os
import re
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv("../keys/.env", override=True)
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=os.getenv("GROQ_API_KEY"))

# ------------------------------------------------------------------
# 1. THE ACTION: look up a fact in a tiny "database" (a Python dictionary)
# ------------------------------------------------------------------
FACTS = {
    "punjab": "The University of the Punjab, Lahore, was founded in 1882.",
    "lums":   "LUMS (Lahore University of Management Sciences) was founded in 1985 by Syed Babar Ali.",
}

def lookup(name):
    """Return the fact whose key appears in the name, e.g. 'LUMS' -> the LUMS fact."""
    for key, fact in FACTS.items():
        if key in name.lower():
            return fact
    return "No information found."

# ------------------------------------------------------------------
# 2. THE INSTRUCTIONS: teach the model the ReAct format, with one example
# ------------------------------------------------------------------
REACT_PROMPT = """Answer the question by repeating Thought, Action and Observation steps.
You can look up facts about two universities: University of the Punjab, and LUMS.
Write ONE Thought and ONE Action, then stop. I will reply with the Observation.
When you know the answer, write a Final Answer instead of an Action.

Example:
Question: When was LUMS founded?
Thought: I need to look up LUMS.
Action: lookup[LUMS]
Observation: LUMS was founded in 1985.
Thought: I now know the answer.
Final Answer: LUMS was founded in 1985."""

# ------------------------------------------------------------------
# 3. THE ReAct LOOP, inside a chat loop
# ------------------------------------------------------------------
while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ("exit", "quit"):
        break

    messages = [{"role": "system", "content": REACT_PROMPT},
                {"role": "user", "content": user_input}]

    for step in range(1, 6):                                    # guardrail: at most 5 passes
        response = client.chat.completions.create(
            model="qwen/qwen3.8-27b",
            messages=messages,
            stop=["Observation:"],                              # stop before the model invents an observation
            reasoning_effort="none",                            # no hidden thinking: the Thought is written out
            max_tokens=300,
        )
        reply = response.choices[0].message.content.strip()
        print(reply)                                            # THINK + ACT (or the final answer)
        messages.append({"role": "assistant", "content": reply})

        match = re.search(r"lookup\[(.*?)\]", reply)            # did the model ask for a lookup?
        if not match:                                           # no action: the model has finished
            break

        observation = lookup(match.group(1))                    # ACT: our code runs the lookup
        print("Observation:", observation)                      # OBSERVE
        messages.append({"role": "user", "content": "Observation: " + observation})
    print()                                                     # a blank line between questions
````

## d. ReAct then (2022) and now

You have just built the **original** ReAct design. Real agents today keep the same loop but change the parts around it:

| | Example 8 (original ReAct, 2022) | Modern agents |
|---|---|---|
| **How the model asks for an action** | It writes text such as `Action: lookup[LUMS]`, and our code finds it with a regular expression | **Native tool calling**: the model returns a structured JSON request *(Lec-40 and Lec-41)* |
| **Where the actions come from** | A Python function over a dictionary | Real APIs, databases and **MCP servers** *(Lec-44 to Lec-47)* |
| **Where the thinking happens** | Written `Thought:` lines in the reply | Often inside a *reasoning* model's hidden thinking, as with `gpt-oss` |
| **Implementation** | A loop you write yourself | Your own loop, or a framework such as the OpenAI Agents SDK *(a later lecture)* |

The **idea is unchanged**: think, act, observe, repeat. Today's models are simply much better at it.

## e. Summary: from a single LLM call to an AI agent

| Example | Design | LLM calls | Who decides the next step? | Components used (Section 3) |
|---|---|---|---|---|
| **1** | A single LLM call | 1 | Nobody | Model, instructions |
| **2** | Chatbot, no memory | 1 per turn | The human | Model, instructions |
| **3, 4, 5** | Chatbot with memory | 1 per turn | The human | + Memory (the context window) |
| **6** | Workflow: prompt chaining | 3, fixed | Your code | Model, instructions, several calls |
| **7** | Workflow: routing | 2 per ticket, fixed | Your code | + a simple guardrail |
| **8** | ReAct agent | As many as needed | **The LLM** | + An action (`lookup`), the ReAct loop, a step limit |

> **The big idea.** Going from Example 1 to Example 8, the model never changed. What changed is the code *around* the model: the memory we send it, the tools we let it use, and the loop that decides when to stop. That code is the **harness**, and it is why people say **Agent = Model + Harness**.

| You get | You pay with |
|---|---|
