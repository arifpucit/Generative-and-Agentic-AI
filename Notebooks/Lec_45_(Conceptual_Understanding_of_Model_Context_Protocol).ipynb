# <span style='background :lightgreen' >5. Overview of Model Context Protocol (MCP)</span>

<h2 align="center"><div class="alert alert-success" style="margin: 20px">MCP (Model Context Protocol) is an open standard created by Anthropic (Nov 2024) that acts as a universal connector using which an agent can connect with external tools, services, and data sources</div></h2>

- **MCP Host:**  An application that contains an MCP client and provides the user interface. The host runs the LLM, manages the UI, and orchestrates communication with MCP servers. It can spawn local servers (running on your machine) or connect to remote servers (hosted on the internet). Some popular MCP Hosts (IDEs & Applications) are:
    - **Claude Desktop:**  A GUI chat interface by Anthropic that support MCP (https://claude.ai/download)
    - **Google Gemini CLI:** A CLI chat interface for Google's Gemini with MCP support
    - **Claude Code:**  AI coding assistant CLI tool
    - **VS Code:** Microsoft's code editor with MCP support (https://code.visualstudio.com/)
    - **GitHub Copilot:**  An AI assistant extension/plugin that runs inside VS Code (and other editors). (https://github.com/features/copilot)
    - **Cursor:** AI-first code editor with native MCP (https://cursor.com/)
    - **Trae:** An AI-powered IDE with built-in MCP client support and a one-click marketplace for MCP servers. (https://www.trae.ai/)
- **MCP Client:**  The software component (connector) embedded in the Host that speaks MCP (JavaScript Object Notation - Remote Procedure Call (JSON-RPC)) to handle communication between MCP host and MCP servers. The MCP client discovers what a server offers, calls its tools, reads its resources and listens for change notifications. A Host can run many clients (one per server or per user context).
    - **What the MCP Client Does:**
        - Sends your requests (like "list files" or "search GitHub") to the appropriate MCP server
        - Receives results from servers and displays them in your application
        - Handles authentication (API keys, OAuth tokens) automatically
        - Manages connections to multiple servers simultaneously (e.g., one for files, one for GitHub)
        - Discovers each server's tools, resources and prompts, runs them, and starts/stops local (stdio) servers
    - **Simple Example:** When you type "What issues are open in my repo?" in Claude Desktop, the MCP Client:
        1. Sends the request to the GitHub MCP Server
        2. Waits for the server to fetch the data
        3. Receives the list of issues
        4. Displays them in the chat interface
- **MCP Server:** An MCP Server is a service (local or remote) that provides tools, resources and prompts to MCP clients. Servers execute the actual work—calling APIs, querying databases, processing files, etc. Some key characteristics are:
    - Each server connects to specific tools, apps, or data sources
    - Designed to be modular (minimal dependencies, easy to add/remove)
    - Can expose multiple tools (e.g., Airbnb server, Slack server).
    - Interfaces with external APIs (Slack, Dropbox, Gmail) or local computations.
- **MCP Protocol:** Standardized communication layer between MCP clients and MCP servers. Defines the transport and data exchange process. The Transport layer of MCP protocol controls how the messages flow between client and server. MCP defines two standard transports: `stdio` for local (spawned) servers and `Streamable HTTP` for remote servers (the older HTTP+SSE transport is deprecated, and WebSocket is not a standard MCP transport). MCP messages are JSON-RPC 2.0 payloads wrapped in the chosen transport. The transport also decides how a connection ends (e.g., closing stdin for a local server).

>- RPC allows a program to execute a function on another computer as if it were local, hiding the details of network communication and data transfer. This abstraction makes it easier to build distributed applications.
>- JSON-RPC combines the concepts of RPC with the simplicity of JSON, allowing developers to structure RPC requests and responses in a standardized JSON format.
>- Unlike REST, JSON-RPC supports batching (multiple requests at once) and notifications (fire-and-forget calls). MCP uses notifications, but removed batching in revision 2025-06-18: every MCP message is a single request, response or notification.
>- JSON-RPC can enable bi-directional communication when used over a persistent connection like WebSocket.

<h3 align="center"><div class="alert alert-success" style="margin: 20px">The MCP Host and Client usually run on local machine (desktop apps), while the MCP servers can run locally as well as on a remote VM-based machine accessible via the network.</div></h3>
<h3 align="center"><div class="alert alert-success" style="margin: 20px">An MCP server—whether running locally or remotely—can handle local computations (e.g., mathematical functions, linear regression, capturing and describing screenshots with an LLM, organizing desktop files, or performing intelligent searches over the local file system) as well as invoke external APIs such as GitHub, Gmail, Dropbox, or Slack.</div></h3>

## a. Local MCP Server (Stdio Protocol)

<img align="right" width="550" src="../images/local-mcpserver.png"  >

- A local MCP server runs on the same machine as the MCP client (i.e., on your local computer) and can access both local resources (files, databases, applications) and remote APIs (Google Drive, GitHub etc).
- Data processing happens on your machine. Require installation and configuration via `claude_desktop_config.json` or similar config files.
- Use stdio transport protocol, when building tools that need to interact with the local environment (e.g., file system, processes, screen automation).
- Typically, each client instance starts its own server process (one per user).
- **How it works:**
    - The MCP client/host is told how to download and run the server (via a config script or package, e.g., `npm`, `npx`, or `uvx`).
    - The MCP client installs and launches the MCP server as a subprocess on the same machine.
    - The client writes JSON-RPC messages to the server’s stdin.
    - The server processes the messages and replies via its stdout. (May or may not access remote services)
- **Benefits:**
    - Very fast communication (direct inter-process communication).
    - More secure — no open network port exposed.
    - Required for tools that access local resources (file system, desktop automation, local search).
    - Easy to set up, no network or extra dependencies.
- **Limitations:**
    - Code updates require each user to reinstall or refresh locally.
    - Computations run on the local machine, which can be resource-heavy.
- **Examples of Local MCP Servers:**
    - Filesystem MCP Server (Access and manage files on your computer): https://github.com/modelcontextprotocol/servers/tree/main/src/filesystem 
    - Git MCP Server (Interact with local Git repositories): https://github.com/modelcontextprotocol/servers/tree/main/src/git
    - GitHub MCP Server (Official server by GitHub; interacts with GitHub APIs using a GitHub PAT): https://github.com/github/github-mcp-server
    - Google Drive MCP Server (Access Google Drive with Google OAuth credentials; archived reference server): https://github.com/modelcontextprotocol/servers-archived/tree/main/src/gdrive
    - Slack MCP Server (Local server for Slack interactions): https://github.com/modelcontextprotocol/servers-archived/tree/main/src/slack (archived), https://github.com/zencoderai/slack-mcp-server
    - Fetch MCP Server (Search the web): https://github.com/modelcontextprotocol/servers/tree/main/src/fetch
    - A Twitter MCP Server: https://github.com/EnesCinr/twitter-mcp

## b. Remote MCP Server (Streamable HTTP Protocol)
<img align="right" width="550" src="../images/remote-mcpserver.png"  >

- A remote MCP server is hosted on the internet rather than your local machine.
- No need to install or run locally, and updates are applied automatically.
- Internet-accessible with simplified sign-in and permission granting.
- Transport protocols for remote servers:
    - **HTTP+SSE** (2024-11-05): the original remote transport. Deprecated since 2025-03-26, so do not use it for new servers.
    - **Streamable HTTP** (2025-03-26 onward): the standard remote transport today. Since revision 2026-07-28 it is **stateless**: every message is its own HTTP POST to one endpoint (e.g., `https://example.com/mcp`) and there is no session ID, so any copy of the server behind a load balancer can answer it.
- **How it works:**
    - The MCP client/host is given the server URL instead of a local install path.
    - The client sends each JSON-RPC request as an HTTP POST to the server's single MCP endpoint, with headers such as `MCP-Protocol-Version: 2026-07-28`, `Mcp-Method: tools/call` and `Mcp-Name: get_weather`.
    - The server replies with either one JSON object or a short Server-Sent Events (SSE) stream that carries progress messages followed by the final result.
    - To be told about changes (e.g., "my tool list changed"), the client keeps one long-lived `subscriptions/listen` request open.
    - Authentication normally uses OAuth 2.1 (some private servers just use an API key in a header). Since 2026-07-28, clients register with the authorization server using Client ID Metadata Documents; the older Dynamic Client Registration is deprecated.
- **Benefits:**
    - Updating the server automatically updates all clients.
    - A single remote server can serve many MCP clients at once.
    - Offloads heavy computations (ML inference, image generation, data processing).
    - More scalable and portable — clients only need the server URL.
    - Enables web-based MCP clients (e.g., ChatGPT or future browser apps).
- **Limitations:**
    - Must be secured (HTTPS, OAuth) because it is reachable over the network.
    - Slightly slower due to network latency.


# <span style='background :lightgreen' >8. Primitives of Model Context Protocol (MCP)</span>

## a. MCP Clients can ask for following Features from MCP Servers

<div style="text-align:center;">
    <img src="../images/mcp-99.png"
         style="max-width:1500px; width:100%; height:auto; display:inline-block;">
</div>

- An MCP server can contain three types of primitives:
    - **Tools** (https://modelcontextprotocol.io/specification/2026-07-28/server/tools) tools are actions (functions, APIs, etc) the AI ask the MCP server to perform. For example a GitHub MCP server might have tools like `createIssue`, `getRepoContent`, `listBranches`, etc.  A Google Drive MCP server might have tools that can search your Google Drive for specific files, or can create new files inside your Google Drive. Similarly, a local MCP server might run local computations (math, regression, screenshot automation). Tools are controlled by the Model (LLM), which decides *when*, *how*, and *with what arguments* to call a tool. Tools are perfect for giving your AI model additional capabilities it can use autonomously. When you ask Claude to "calculate the square root of 3 using JavaScript," it's Claude that decides to use a JavaScript execution tool to run the calculation. The key property of a tool is that it performs an action (not just data retrieval). In MCP, each tool must define Name (e.g., `get_weather`), Description (explains what the tool does and the arguments it expects) and Input Schema (dictionary of arguments the tool accepts). Optionally, a tool can also declare an Output Schema (so its result comes back as structured JSON), annotations (e.g., read-only or destructive) and icons. To deal with tools, MCP server normally provides two functions:
        - `tools/list` MCP client ask the MCP server "What tools do you provide?"
        - `tools/call` MCP client tells the MCP server "Please run this tool with these arguments"
    - **Resources:** (https://modelcontextprotocol.io/specification/2026-07-28/server/resources) Resources are structured data sources (static context and data in string, JSON or binary format), for the user or the AI model to use. Resources are controlled by your application code. Your app decides when to fetch resource data and how to use it - typically to add context to conversations. For example, the MCP host/client if connected to GitHub MCP server can request it to fetch the README file of a specific GitHub repository. Similarly, if the MCP client if connected to some MCP Database server can request it to fetch the database schema of a specific database. The purpose of a resource is that it supplies background data (**context**) rather than executing actions. (A resource can be a direct resource or templated resource). To deal with resources, MCP server normally provides following functions:
        - `resources/list` MCP client ask the MCP server "What resources are available?"
        - `resources/templates/list` MCP client ask the MCP server "Which resource templates (URIs with parameters, e.g., `file:///{path}`) do you offer?"
        - `resources/read` MCP client ask the MCP server "Give me the content of this resource"
        - `subscriptions/listen` MCP client asks to be told when a resource (or the list of tools, prompts or resources) changes. Since 2026-07-28 this single request replaces `resources/subscribe` and `resources/unsubscribe`; to stop listening, the client cancels it.
    - **Prompt Templates:** (https://modelcontextprotocol.io/specification/2026-07-28/server/prompts) Prompts are predefined prompt templates that the MCP server offers that client/model can use. The purpose of a prompt is that it helps users apply effective, pre-crafted LLM instructions. For example, scientific research report templates, literature review prompts and report-writing scaffolds. The workflow of using prompt is first the user selects a prompt which replaces/augments the user's input and the LLM is called with that structured prompt for better results. To deal with prompts, MCP server normally provides two functions:
        - `prompts/list` MCP client ask the MCP server "What prompt templates do you provide?"
        - `prompts/get` MCP client fetches a specific prompt template

> Since 2026-07-28, list and read results also carry a cache hint (`ttlMs`), so a client can reuse a tool list for a while instead of asking for it again before every request.

## b. MCP Servers can ask for following Features from MCP Clients
<div style="text-align:center;">
    <img src="../images/mcp-new-29.png"
         style="max-width:1500px; width:100%; height:auto; display:inline-block;">
</div>

- While working on a request, a server sometimes needs help from the client. MCP defines three such features. Revision 2026-07-28 **deprecated two of them** (Sampling and Roots): they still work, but new servers should not use them, and they can be removed in the first revision released on or after 2027-07-28.
- **Sampling** *(deprecated)*: (https://modelcontextprotocol.io/specification/2026-07-28/client/sampling) MCP server asks the client to ask the AI model for its intelligence. An example of a **GitHub MCP server** can be:
```
Server → Client: "I found 50 changed files. Can you ask the AI to write a good commit message summarizing these changes?"
Client → AI Model: "Write a commit message for these changes..."
AI Model → Client: "commit message....."
Client → Server: Here's what the AI said: "commit message..."
```
> **Use instead:** the server calls an LLM provider's API (e.g., OpenAI, Groq) itself.

- **Roots** *(deprecated)*: (https://modelcontextprotocol.io/specification/2026-07-28/client/roots) MCP server asks the client which folders (filesystem boundaries) it may work in. An example of a **File MCP server** can be:
```
Server → Client: "Where am I allowed to work? Which folders can I access?"
Client → Server: "You can access: /home/user/projects/ and /home/user/documents/"
Server: (Now knows it can only read/write files in those two locations)
```
> **Use instead:** pass the folders as tool arguments, as resource URIs, or in the server's configuration (e.g., the folder list given to the server in `claude_desktop_config.json`).

- **Elicitation** *(active)*: (https://modelcontextprotocol.io/specification/2026-07-28/client/elicitation) MCP server asks the client to get some additional information from the human user. It has two modes:
    - **Form mode:** the client shows a small form and sends the answer back to the server. Use it for ordinary data such as a name, a choice or a confirmation.
    - **URL mode** (added in 2025-11-25): the client opens a page of the server in the browser, and the user types the information there. Use it for secrets such as passwords, API keys, OAuth sign-in and payments. The spec **forbids** asking for these in form mode, because the secret would pass through the client (and could reach the LLM).

An example of a **Database MCP server** using form mode:
```
Server → Client: "Which database should I query: dev or prod?"
Client → User: shows a form with two choices
User picks: "dev"
Client → Server: {"action": "accept", "content": {"database": "dev"}}
```
The same server using URL mode to get the password:
```
Server → Client: "Please enter your database password on this secure page: https://db-server.example.com/connect"
Client → User: "The database server wants to open db-server.example.com. Allow?"
User clicks "Allow" and types the password on that page (the client never sees it)
Client → Server: sends the original request again; the server now has the password
```

### How the server asks in 2026-07-28: Multi Round-Trip Requests (MRTR)
#### https://modelcontextprotocol.io/specification/2026-07-28/basic/patterns/mrtr
- **Before 2026-07-28**, the server sent its own JSON-RPC request (e.g., `elicitation/create`) to the client in the middle of a tool call. That needs an open, stateful connection.
- **Now the server answers with a question.** Its reply has `resultType: "input_required"`, the questions in `inputRequests`, and an opaque `requestState` string. The client collects the answers and sends the **same request again** (with a new id), adding `inputResponses` and the unchanged `requestState`. Only `tools/call`, `resources/read` and `prompts/get` can be answered this way.

The server's reply in step 2 of the figure (the question is inside `inputRequests`):
```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "result": {
    "resultType": "input_required",
    "inputRequests": {
      "pick_db": {
        "method": "elicitation/create",
        "params": {
          "mode": "form",
          "message": "Which database should I query?",
          "requestedSchema": {
            "type": "object",
            "properties": {
              "database": {"type": "string", "enum": ["dev", "prod"]}
            },
            "required": ["database"]
          }
        }
      }
    },
    "requestState": "opaque-string-from-the-server"
  }
}
```
The client's retry in step 4 (`_meta` left out for brevity):
```json
{
  "jsonrpc": "2.0",
  "id": 2,
  "method": "tools/call",
  "params": {
    "name": "run_query",
    "arguments": {"sql": "SELECT COUNT(*) FROM users"},
    "inputResponses": {
      "pick_db": {"action": "accept", "content": {"database": "dev"}}
    },
    "requestState": "opaque-string-from-the-server"
  }
}
```

# <span style='background :lightgreen' >10. MCP Communication Life Cycle</span>
#### https://modelcontextprotocol.io/specification/2026-07-28/basic/versioning
- The way a client and a server start, run and end their conversation changed in revision **2026-07-28**.
- You will meet both styles: many servers, hosts and tutorials still use the older one, and the official SDKs (e.g., the Python `mcp` package, version 2.x) speak both.

## a. Older Revisions (2025-11-25 and earlier): Initialize → Operate → Shutdown
<div style="text-align:center;">
    <img src="../images/mcp-new-27.png"
         style="max-width:1500px; width:100%; height:auto; display:inline-block;">
</div>

### (i) Initialization Phase
- **Initialization phase** is the starting handshake between the **client** and the **server**. Its goal is to make sure both sides agree on how to talk (protocol version) and what features (capabilities) are available.
    - **Client → Server: Initialization Request** The client sends an `initialize` request that includes the protocol version it wants, its capabilities (features it supports) and client info (name, version).
    - **Server → Client: Initialization Response** The server responds with the protocol version it will use, its capabilities and server info (name, version).
    - **Agreement on Protocol Version** If the server supports the client's version, it replies with the same version (e.g., `2025-11-25`). Otherwise it replies with the latest version it supports, and the client disconnects if it cannot speak that version.
    - **Agreement on Capabilities** Client and server check what features both support. Only the common features are used in later communication.
    - **Client → Server: Initialized Notification** The client sends a final `notifications/initialized` message. This means: *“I got your info, I’m ready to start normal operations.”*

### (ii) Operation Phase
- **Operation** is the main working phase, where the client and server exchange real data and requests
    - Client requests → list tools, call tools, read resources, list and get prompts.
    - Server responses → execute tools, return results.
    - Server requests → sampling, roots and elicitation, sent by the server in the middle of a client request (Section 8b).
    - Notifications → progress updates, logs, and changes such as `notifications/tools/list_changed` or `notifications/resources/updated`.
    - Example: Claude client asks GitHub MCP server for repository file contents.

### (iii) Shutdown Phase
- **Shutdown** is the final stage, when the session ends, either normally or due to an error/timeout. MCP defines **no shutdown message**: one side (usually the client) simply closes the transport.
    - **stdio:** the client closes the server's stdin, waits for the process to exit, and sends `SIGTERM` (then `SIGKILL`) if it does not exit in time.
    - **Streamable HTTP:** the client closes the HTTP connection(s). It can also end the session explicitly by sending an HTTP `DELETE` with the `Mcp-Session-Id` header.
    - Before closing, ongoing requests should be completed or cancelled (`notifications/cancelled`).

## b. Latest Revision (2026-07-28): Stateless Requests
<div style="text-align:center;">
    <img src="../images/mcp-new-28.png"
         style="max-width:1500px; width:100%; height:auto; display:inline-block;">
</div>

- **No handshake:** `initialize` and `notifications/initialized` are gone. Instead, every request carries three fields inside `params._meta`:
    - `io.modelcontextprotocol/protocolVersion`: the version of this request, e.g., `"2026-07-28"`
    - `io.modelcontextprotocol/clientCapabilities`: what the client supports, e.g., `{"elicitation": {}}`
    - `io.modelcontextprotocol/clientInfo` (recommended): who is asking, e.g., `{"name": "ExampleClient", "version": "1.0.0"}`
- **No session:** the server remembers nothing between requests, so each request can be answered by a different copy of the server behind a load balancer. If a tool needs to remember something across calls (e.g., a shopping basket), the server returns an ID (e.g., `basket_id`) and the model passes it back as an ordinary tool argument in the next call.
- **Discovery (optional):** the client may first call `server/discover` to get the server's supported versions, capabilities and name in one request. Every server must implement it.
- **Version mismatch:** if the server does not support the requested version, it returns an `UnsupportedProtocolVersionError` listing the versions it does support, and the client retries with one of them.
- **Result type:** every result now has a `resultType`: `"complete"` for a normal answer, or `"input_required"` when the server needs more input first (Section 8b).
- **Ending:** there is still no shutdown message. With stdio, the client closes the server's stdin (then `SIGTERM`/`SIGKILL` if needed). With Streamable HTTP there is nothing to close (apart from an open `subscriptions/listen` stream), because each request is a separate HTTP POST.
- **Old and new together:** a client that supports both first tries the new way (on stdio by sending `server/discover`); if the server does not understand it, the client falls back to `initialize`.

# <span style='background :lightgreen' >11. What's New in MCP 2026-07-28 ("MCP 2.0")</span>
#### https://modelcontextprotocol.io/specification/2026-07-28/changelog
<h3 align="center"><div class="alert alert-success" style="margin: 20px">MCP versions are dates, not numbers. "MCP 2.0" is the community's name for revision 2026-07-28, the biggest change since MCP launched. The official SDKs moved to version 2 to implement it (e.g., the Python <code>mcp</code> package 2.0.0, released on 28 July 2026).</div></h3>

### a. What Changed
| Topic | Before (2025-11-25 and earlier) | Now (2026-07-28) |
|---|---|---|
| Starting a conversation | `initialize` + `notifications/initialized` handshake | No handshake: version, capabilities and client info travel in every request's `_meta` |
| Sessions | One session per connection (`Mcp-Session-Id` header on HTTP) | No sessions: any copy of the server can answer any request |
| Learning what a server supports | The `initialize` response | `server/discover` (optional for the client) |
| Server needs input from the client | Server sends its own request (`sampling/createMessage`, `roots/list`, `elicitation/create`) | Server replies `input_required`; the client retries with `inputResponses` (MRTR) |
| Change notifications | `resources/subscribe` / `resources/unsubscribe` and an HTTP GET stream | One `subscriptions/listen` request |
| Results | No result type | Every result has a `resultType`: `complete` or `input_required` |
| Caching | Not defined | List and read results carry `ttlMs` and `cacheScope` |
| Health check and logging | `ping`, `logging/setLevel` | `ping` removed; the log level is set per request in `_meta` |
| Long-running jobs | Tasks (experimental, part of the core) | Tasks moved to an official extension |
| HTTP headers | `MCP-Protocol-Version`, `Mcp-Session-Id` | `MCP-Protocol-Version`, `Mcp-Method`, `Mcp-Name` |
| Broken HTTP stream | Resumed with `Last-Event-ID` | No resume: the client sends the request again with a new id |

### b. Deprecated Features (they still work, but new code should not use them)
| Feature | Use instead | Earliest removal |
|---|---|---|
| Sampling | Call an LLM provider's API directly | First revision released on or after 2027-07-28 |
| Roots | Pass folders as tool arguments, resource URIs or server configuration | First revision released on or after 2027-07-28 |
| Logging | Write logs to `stderr` (stdio) or use OpenTelemetry | First revision released on or after 2027-07-28 |
| OAuth Dynamic Client Registration | Client ID Metadata Documents | First revision released on or after 2027-07-28 |
| HTTP+SSE transport (deprecated since 2025-03-26) | Streamable HTTP | Three months after the deprecation policy (SEP-2596) is final |

### c. Official Extensions (optional add-ons, off by default; both sides advertise them in their capabilities)
| Extension | What it adds | Stage |
|---|---|---|
| MCP Apps | Servers show interactive UI (charts, forms, video players) inside the chat | Stable |
| Tasks | Long-running jobs: start now, check for the result later | Beta (planned to move into the core) |
| Skills over MCP | Servers share Agent Skills (instructions and files) through resources | Beta |
| Server Cards | Clients read a server's metadata before connecting to it | Beta |
| OAuth Client Credentials | Machine-to-machine sign-in | Stable |
| Enterprise-Managed Authorization | Central, company-managed access control | Stable |

# <span style='background :lightgreen' >12. Timeline of MCP Specification Revisions</span>

| Revision | Main changes |
|---|---|
| **2024-11-05** (first; made public by Anthropic on 25 Nov 2024) | Tools, resources, prompts, sampling, roots; stdio and HTTP+SSE transports |
| **2025-03-26** | Streamable HTTP replaces HTTP+SSE; OAuth 2.1 authorization; tool annotations (e.g., read-only, destructive); audio content; JSON-RPC batching |
| **2025-06-18** | Elicitation; structured tool output (`outputSchema`); resource links in tool results; JSON-RPC batching removed; `MCP-Protocol-Version` header required on HTTP |
| **2025-11-25** | Tasks (experimental); URL-mode elicitation; icons; tool calling inside sampling; Client ID Metadata Documents |
| **Dec 2025** (governance) | Anthropic donates MCP to the Agentic AI Foundation (Linux Foundation) |
| **2026-07-28** ("MCP 2.0") | Stateless requests, `server/discover`, MRTR, `subscriptions/listen`, caching hints, official extensions; Sampling, Roots and Logging deprecated |
