## g. Running Claude Desktop with Ollama

### Overview
- Claude Desktop can be integrated with Ollama Cloud, enabling the use of Ollama-hosted models — including within Claude Cowork and Claude Code inside the app. This is particularly useful for students and practitioners who wish to leverage powerful third-party models as alternatives to Anthropic's native models (Opus, Sonnet, Haiku), often at significantly lower cost or entirely free of charge.
- **Important Note on Claude Cowork and Claude Code:** These two features within Claude Desktop are available to all users once the Ollama integration is configured. You do **not** necessarily need to upgrade to a paid Anthropic subscription plan to use them via Ollama — the integration routes inference through Ollama's infrastructure rather than Anthropic's API.
- If you are on a budget or wish to experiment with open-weight models, using Claude Desktop via Ollama Cloud is an excellent approach. Many Ollama Cloud models are available on a **free tier with generous limits**, making this setup practical for coursework and personal projects without incurring API charges.

### Prerequisites
- Before proceeding, ensure the following requirements are met:
    - **Claude Desktop** must be installed on your machine (available for **macOS** or **Windows**).
    - **Ollama** must be installed and updated to at least **v0.15+** (the `ollama launch` command was introduced in this version). You can download the latest version from [https://ollama.com/download](https://ollama.com/download).
    - An Ollama API Key** is required. Obtain yours from [https://ollama.com/settings/keys](https://ollama.com/settings/keys), then set it in your terminal shell **before** launching:
```bash
# On macOS/Linux
export OLLAMA_API_KEY=your_api_key_here

# On Windows (PowerShell)
$env:OLLAMA_API_KEY="your_api_key_here"
```

### Step-by-Step Setup

#### Step 1 — Configure Claude Desktop to Use Ollama
- Run the following command in your terminal. This performs the **quick setup** that configures Claude Desktop to use Ollama Cloud as its inference backend. The following command automatically modifies Claude Desktop's configuration so that it points to Ollama Cloud instead of Anthropic's API. No manual editing of environment variables or configuration files is needed.

```bash
$ ollama launch claude-desktop
```

#### Step 2 — Explore Available Ollama Cloud Models
- After setup, Claude Desktop **automatically discovers** your available Ollama Cloud models. For example, `kimi-k2.6` will appear inside Claude Cowork, and the same models are also available to Claude Code inside Claude Desktop.
- Some of the notable Ollama Cloud models you may encounter include:
| Model | Provider | Notes |
|---|---|---|
| `kimi-k2.6` | Moonshot AI | Strong reasoning and coding |
| `qwen3.5` | Alibaba | Multilingual, instruction-tuned |
| `minimax-m2.7` | MiniMax | Multimodal capabilities |
| `glm-5.1` | Zhipu AI | Efficient general-purpose model |

> You can browse the full model library at [https://ollama.com/library](https://ollama.com/library).

### Additional Commands

#### Configure Without Launching
- If you want to apply the configuration changes **without immediately launching** Claude Desktop, use:
```bash
$ ollama launch claude-desktop --config
```

#### Restore the Default Anthropic Profile
To switch Claude Desktop back to its **standard Anthropic profile** (using native Claude models such as Opus, Sonnet, and Haiku), run:
```bash
$ ollama launch claude-desktop --restore
```
- If Claude Desktop is already running and you wish to apply the restore without manual intervention, append the `--yes` flag to automatically approve the restart:
```bash
$ ollama launch claude-desktop --restore --yes
```

### What Is Currently Supported (and What Is Not)

    - **Ollama Cloud** as the third-party inference gateway

# <span style='background :lightgreen' >3. Extending Claude Desktop with MCP Servers</span>

## a. Ways to Connect your Claude Desktop to a Tool
- Claude Desktop acts as an MCP client, so it can connect to local and remote MCP servers to extend Claude's capabilities with external tools and data.
- By default, no external tools are connected. You can add them in three ways:
  | Method | Protocol | Plan required | Best for |
  |---|---|---|---|
  | **Connectors** (Settings → Connectors) | Streamable HTTP | Free (1 connector max); unlimited on Pro+ | Remote MCP servers (GitHub, Atlassian, HuggingFace etc.) |
  | **Extensions** (Settings → Extensions) | stdio | Free and above | Local MCP servers via one-click `.mcpb` bundle install |
  | **Config file** (`claude_desktop_config.json`) | stdio | Free and above | Custom/private local servers (full manual control) |

> **Free-tier students:** You can use Extensions and the config file to connect local MCP servers, and you get one free custom Connector for a remote MCP server. Unlimited directory Connectors require a paid plan (Pro or above).


## a. Ways to Connect Claude Desktop to a Tool

- Claude Desktop acts as an **MCP client**, so it can connect to both local and remote MCP servers to extend Claude's capabilities with external tools, files, and data sources.
- By default, no external tools are connected. There are four distinct methods to add them:

| Method | Where to configure | Transport | Plan required | Best for |
|---|---|---|---|---|
| **[Directory Connectors](https://support.claude.com/en/articles/11176164-use-connectors-to-extend-claude-s-capabilities)** | Settings → Connectors | Remote MCP (HTTP/SSE) | Free (some); unlimited on Pro+ | Pre-built, Anthropic-verified cloud services (GitHub, Atlassian, Slack, Google Drive, HuggingFace, etc.) |
| **[Custom Connectors](https://support.claude.com/en/articles/11175166-get-started-with-custom-connectors-using-remote-mcp)** | Settings → Connectors → + | Remote MCP (HTTP/SSE) | Free (1 max); unlimited on Pro+ | Your own self-hosted remote MCP server, reachable over the public internet |
| **[Desktop Extensions](https://support.claude.com/en/articles/10949351-getting-started-with-local-mcp-servers-on-claude-desktop)** | Settings → Extensions | stdio (local) | Free and above | Local MCP servers via one-click `.mcpb` or `.dxt` bundle install from the directory or a custom file |
| **[Config file](https://docs.claude.com/en/docs/claude-code/mcp)** (`claude_desktop_config.json`) | Manual file edit | stdio (local) | Free and above | Custom/private local MCP servers requiring full manual control over arguments and environment variables |

> **Important corrections to the original description:**
> - **[Directory Connectors](https://claude.ai/directory/connectors)** (the pre-built ones such as Slack, Google Drive, etc.) are **not** limited to one on the Free plan — many directory connectors are available to Free users. The **one-connector limit applies specifically to *custom* connectors** (i.e., self-hosted remote MCP servers you add manually).
> - **Connectors and Desktop Extensions are two distinct mechanisms.** Connectors (both directory and custom) connect Claude to **remote, internet-hosted** MCP servers via HTTP/SSE from Anthropic's cloud infrastructure. Desktop Extensions connect Claude to **locally running** MCP servers on your own machine via stdio. They are configured in different places in Settings.
> - The `claude_desktop_config.json` config file method is a **separate, lower-level alternative** to Desktop Extensions for local MCP servers — it does not apply to remote connectors.

> **Free-tier students:** You have access to many pre-built directory connectors and unlimited desktop extensions (local). You may also add **one** custom (self-hosted) remote MCP connector. Your server must be publicly reachable on the internet for custom connectors to work — Anthropic's cloud brokers the connection, not your local machine. For fully local/private MCP servers, use Desktop Extensions or the config file instead.

# <span style='background:lightgreen'>4. Hands-on Practice of Connecting Claude Desktop with MCP Servers using **Directory Connectors**</span>

- Directory Connectors connect Claude to **cloud-hosted (remote) MCP servers** over the internet using the **Streamable HTTP** transport (the older HTTP+SSE transport is being phased out as of 2025).
- If a tool talks to a cloud API — such as Slack, GitHub, or Stripe — it is a remote Connector, not a Desktop Extension.
- Once connected, remote Connectors are available **everywhere** — Claude Desktop, Claude web (claude.ai), Claude mobile (Android/iOS), Cowork, and Claude Code — without any additional setup.
- **Plan requirements:**
  - **Directory Connectors** (pre-built, Anthropic-verified): available to **Free and paid users** alike.
  - **Custom Connectors** (self-hosted remote MCP servers you add manually): available to all plans, but **Free users are limited to one** custom connector; paid plans (Pro and above) have unlimited custom connectors.
- Authentication is handled automatically by Claude via **OAuth** (e.g., sign in with Google or GitHub) or **API keys** — no manual configuration required.
- Ideal for non-technical users: connectors are easy to add and remove through the Claude UI at **Settings → Connectors** or by browsing [claude.com/connectors](https://claude.com/connectors).

# <span style='background:lightgreen'>6. Hands-on Practice of Connecting Claude Desktop with MCP Servers using **`claude_desktop_config.json`**</span>

- Not all MCP servers have a Connector or Extension wrapper built for them. For servers without one-click support — including any custom or private local MCP server you build yourself — you register them manually in the `claude_desktop_config.json` file.
- **Important distinction:** Local MCP servers configured via this file use **your local machine's network** directly. This is a fundamentally different mechanism from remote Connectors (which route through Anthropic's cloud). Consequently, servers registered here are **not available** in Cowork or claude.ai — only in Claude Desktop and Claude Code.

**Step 1 — Locate and open the config file:**

| OS | File path |
|---|---|
| macOS | `~/Library/Application Support/Claude/claude_desktop_config.json` |
| Windows | `%APPDATA%\Claude\claude_desktop_config.json` |

The most reliable way to open it is directly from within the app:
- **macOS:** Claude menu (top system menu bar) → **Settings → Developer → Edit Config**
- **Windows:** Inside the app → **Settings → Developer → Edit Config**

This opens (or creates) the config file in your default text editor.

**Step 2 — Add your MCP server configuration:**

Obtain the MCP server's configuration snippet from its documentation (GitHub page, npm package page, etc.) and paste it into the `mcpServers` object. Below is a complete example registering the official Filesystem MCP server:

```json
{
  "mcpServers": {
    "filesystem": {
      "command": "npx",
      "args": ["-y", "@modelcontextprotocol/server-filesystem", "/Users/yourname/Documents"]
    }
  }
}
```

Save the file after pasting.

**Step 3 — Fully quit and relaunch Claude Desktop:**

Claude Desktop reads the config file **only at startup**, so changes will not take effect until you restart the application completely:

- **macOS:** Claude menu (top system menu bar) → **Quit**, or press **Cmd+Q**, then relaunch
- **Windows:** Right-click the Claude icon in the system tray → **Quit**, or use Task Manager → End Task, then relaunch

**Step 4 — Verify the integration:**

After relaunching, click the **`+`** button at the bottom of the chat input box → **Connectors**. The tools exposed by your newly registered MCP server should appear in the list. If they do, the server is connected and fully operational.

### Step 6: Extract Client ID and Client Secret

Open the `oauth_credentials.json` file. It will look like this:

```json
{
  "installed": {
    "client_id": "your-client-id.apps.googleusercontent.com",
    "project_id": "claude-mcp-xxxxx",                        #this needs to be updated
    "auth_uri": "https://accounts.google.com/o/oauth2/auth",
    "token_uri": "https://oauth2.googleapis.com/token",
    "auth_provider_x509_cert_url": "https://www.googleapis.com/oauth2/v1/certs",
    "client_secret": "your-client-secret",                  #this need to be updated
    "redirect_uris": ["http://localhost"]
  }
}

```

**Copy these two values:**
- `client_id`: The value that ends with `.apps.googleusercontent.com`
- `client_secret`: The secret string

You'll need these for your Claude Desktop configuration.

### Step 7: Configure Claude Desktop

Edit your `claude_desktop_config.json`:

```bash
open ~/Library/Application\ Support/Claude/claude_desktop_config.json
```

Add the Google Drive MCP server configuration:

```json
{
  "mcpServers": {
    "gdrive": {
      "command": "npx",
      "args": [
        "-y",
        "@modelcontextprotocol/server-gdrive"
      ],
      "env": {
        "GDRIVE_CLIENT_ID": "your-client-id.apps.googleusercontent.com",
        "GDRIVE_CLIENT_SECRET": "your-client-secret"
      }
    }
  }
}
```

**Replace:**
- `your-client-id.apps.googleusercontent.com` with your actual Client ID
- `your-client-secret` with your actual Client Secret

## Troubleshooting

**Issue 1: "OAuth consent screen not configured"**
- Make sure you completed Step 3 (Configure OAuth Consent Screen)
- Add yourself as a test user

**Issue 2: "Access blocked: This app's request is invalid"**
- Check that Google Drive API is enabled (Step 2)
- Verify your redirect URIs in the OAuth client settings

**Issue 3: "Error 403: access_denied"**
- Make sure your email is added as a test user in the OAuth consent screen
- Check that you're signing in with the correct Google account

**Issue 4: Browser doesn't open for authentication**
- Check Claude Desktop logs: `tail -f ~/Library/Logs/Claude/mcp-server-gdrive.log`
- Try manually running: `npx -y @modelcontextprotocol/server-gdrive`

**Issue 5: "Invalid client"**
- Double-check your Client ID and Client Secret are copied correctly
- Ensure there are no extra spaces or line breaks
- Make sure you're using the credentials from a "Desktop app" OAuth client

- `npx` automatically downloads and runs the latest version of the filesystem server.  No manual installation required.  Always gets updates automatically. No maintenance needed

```python
{
      "mcpServers": {               # Container for all MCP servers
        "filesystem": {             # Server name (you can choose any name)
          "command": "npx",         # Command to run (npx executes npm packages)
          "args": [                 # Arguments passed to the command
            "-y",                   # Auto-confirm package installation
            "@modelcontextprotocol/server-filesystem",  # The MCP filesystem server package
            "/path/to/directory1"    # Directory path to grant access
            "/path/to/directory2"    # Directory path to grant access
          ]
        }
      },
    
    "gdrive": {
      "command": "npx",
      "args": [
        "-y",
        "@modelcontextprotocol/server-gdrive"
      ],
      "env": {
        "GDRIVE_CLIENT_ID": "your-client-id.apps.googleusercontent.com", # Copy-Paste the client_id from your downloaded oauth_credentials.json file
        "GDRIVE_CLIENT_SECRET": "your-client-secret"                     # Copy-Paste the client_secret from your downloaded oauth_credentials.json file
      }
    }
}
```
