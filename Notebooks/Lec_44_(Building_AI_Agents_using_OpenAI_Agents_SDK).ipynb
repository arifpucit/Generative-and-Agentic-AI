## a. What is an Agentic Framework?

An **Agentic Framework** is a pre-built structure that provides patterns, abstractions, and orchestration logic specifically designed for building autonomous AI agents.
</div>

**Key Characteristics:**
- **Built-in orchestration:** Automatically handles the agent's reasoning loop
- **Tool management:** Pre-built patterns for registering and executing tools
- **State & memory:** Built-in systems for maintaining context across interactions
- **Multi-agent coordination:** Patterns for multiple agents working together
- **Error handling:** Automatic retry logic and graceful degradation

**Real-world analogy:** Think of it like a car manufacturing assembly line. Instead of hand-crafting every component from scratch, the framework provides the conveyor belts, robotic arms, and quality control stations, you just define what car you want to build.

## b. What Problems Do Agentic Frameworks Solve?

Building agentic systems involves complex challenges:

1. **The Orchestration Problem:** How do you create the loop where the agent thinks → acts → observes → thinks again?
2. **The Tool Calling Problem:** How do you reliably execute functions and feed results back to the agent?
3. **The State Problem:** How do you maintain conversation history and context?
4. **The Multi-Agent Problem:** How do you coordinate multiple specialized agents working on the same task?
5. **The Error Handling Problem:** What happens when a tool fails or the agent gets stuck?

**Frameworks solve these problems** by providing pre-built components and patterns, so you can focus on your specific use case rather than reinventing the wheel.

## c. Core Components of Agentic Frameworks
- **Agent Definitions:** Define the agent's role, capabilities, and behavior
- **Tool/Function Registry:** A system for registering Python functions that agents can call
- **Orchestration Engine:** The "brain" that manages the agent loop: reasoning → tool calling → response

- **Memory Management:** Systems for storing and retrieving conversation history and state
- **Multi-Agent Coordination (Architectural Patterns):** Patterns for agent handoffs, hierarchies, and collaboration

## d. Popular Agentic Frameworks
### (i) LangChain
- **Official Documentation: [https://python.langchain.com/docs/get_started/introduction](https://python.langchain.com/docs/get_started/introduction)**
- LangChain is open source and is introduced in October 2022
- A general LLM app framework (chains, tools, RAG) that also supports basic agents.
- Massive ecosystem: connectors, memory systems, retrievers, deployment helpers.
- Not primarily an agent orchestration engine — that role is now complemented by LangGraph.
- Good for teaching LLM app foundations (RAG, tools + memory), however, is not ideal for complex agents by itself.

### (ii) LangGraph
- **Official Documentation: [https://langchain-ai.github.io/langgraph/](https://langchain-ai.github.io/langgraph/)**
- LangGraph is open-source and is introduced in May 2023 (as part of LangChain evolution).
- A graph-orchestration layer for stateful, conditional agent workflows.
- Each step/task is a node; edges capture transitions and conditions.
- Supports long-running agents, streaming, human-in-loop, checkpoints.
- Good for systems with complex branching, loops, and durable state, where simple scripts are insufficient.

### (iii) AutoGen
**Official Documentation: [https://microsoft.github.io/autogen/](https://microsoft.github.io/autogen/)**
- AutoGen is open-source and is introduced in June 2023.
- AutoGen is a multi-agent conversation & collaboration framework that is designed for agents that converse with each other to accomplish tasks.
- Agents can call tools, involve humans, and exchange messages in a loop.
- Good for agent systems where dialogue and agent interaction patterns matter (e.g., sequential discussions, negotiation bots).

### (iv) CrewAI
- **Official Documentation: [https://docs.crewai.com/](https://docs.crewai.com/)**
- CrewAI is open-source and is introduced in early 2024.
- It is a multi-agent orchestration framework based on the crew/team metaphor.
- Agents are assigned roles, goals, and backstories and work collaboratively to complete tasks.
- Good for projects that benefit from role-based collaboration and intuitive multi-agent delegation.

### (v) OpenAI Agents SDK
**Official Documentation: [https://openai.github.io/openai-agents-python/](https://openai.github.io/openai-agents-python/)**
- Agents  SDK is also Open-Source and introduced in March 2025.
- Lightweight, Python-first framework for building multi-agent workflows with core primitives (Agents, Handoffs, Guardrails).
- Tight integration with OpenAI models & Responses API for decision-making and tool execution.
- Excellent for rapid prototyping and production agents without much overhead.
- Good for projects where OpenAI models are primary and you want simple, fast agent orchestration with low learning overhead.

## e. Design Patterns for Building Effective Agents
- https://www.anthropic.com/engineering/building-effective-agents
  
<img align=right src="../images/Agents000.png" width="900">

### (i) Chaining
- This pattern involves breaking down complext tasks into a sequence of simpler, manageable steps, where each step builds upon the previous one
- The Agents SDK supports this by allowing developers to define agents that execute specific functions in a predetermined order, ensuring a structured approach to task completion

### (ii) Routing
- Routing entails directing tasks to the most appropriate agent based on the task's nature.
- The Agents SDK facilitates this through its handoff mechanism, enabling agents to transfer control to other agents better suited to handle specific subtasks, thereby optimizing task management


### (iii) Parallelization
- This pattern focusses on executing multiple subtasks concurrently to enhance efficiency.
- With Agents SDK developers can design agents that operate in parallel, leveraging the SDK's orchestration capabilities to manage simultaneous processes effectively.

<img align=right src="../images/agents001.png" width="900">

### (iv) Orchestration-Workers
- In this design an orchestrator agent decomposes a complex task into smaller subtasks and assigns them to worker agents.
- The Agent's SDK architecture supports this by allowing an orchestrator agent to oversee the workflow and delegate tasks to specialized worker agents, ensuring coordinated task execution.

### (v) Evaluator
- This pattern involves iterative improvement through feedback loops, where an evaluator agent assesses the performance of other agents and suggest optimizations.
- The Agent's SDK guardrails feature enables the implementation of such evaluative mechanisms, allowing for continuous performance enhancement and adherence to desired behaviors.

# <span style='background :lightgreen' >4. OpenAI Agents SDK Framework</span>

## Main features
- **Agent loop:** A built-in agent loop that handles tool invocation, sends results back to the LLM, and continues until the task is complete.
- **Python-first:** Use built-in language features to orchestrate and chain agents, rather than needing to learn new abstractions.
- **Agents as tools / Handoffs:** A powerful mechanism for coordinating and delegating work across multiple agents. (e.g., transferring control to a specialized agent)
- **Input/Output Guardrails:** Safety measures that validate inputs and outputs to ensure appropriate behavior.
- **Function tools:** Turn any Python function into a tool with automatic schema generation and Pydantic-powered validation.
- **MCP server tool calling:** Built-in MCP server tool integration that works the same way as function tools.
- **Sessions:** A persistent memory layer for maintaining working context within an agent loop.
- **Human in the loop:** Built-in mechanisms for involving humans across agent runs.
- **Tracing:** Built-in tracing for visualizing, debugging, and monitoring workflows, with support for the OpenAI suite of evaluation, fine-tuning, and distillation tools.
- **Realtime Agents:** Build powerful voice agents with features such as automatic interruption detection, context management, guardrails, and more.

### Running the Agent
- Once you’ve created an agent, you can run it using the `Runner` class. This class handles the execution of agent tasks and manages the conversation flow:
- OpenAI gives us three methods for running our agent, all via a Runner class — those methods are:
    - `Runner.run()` which runs your agent asynchronously, meaning it returns a coroutine. You need to await it in an async function. It handles the entire task execution and returns the final result once the agent finishes.
    - `Runner.run_sync()` which runs the agent synchronously, blocking until the task is complete. It’s easier to use in standard Python scripts or notebooks that are not async.
    - `Runner.run_streamed()` which runs the agent asynchronously and streams the response in real-time, token by token or chunk by chunk. You can process or display each piece as it comes in.
- The OpenAI Agents SDK uses asynchronous programming extensively.
- **Documentation Link: https://openai.github.io/openai-agents-python/running_agents/**

````python
# The Runner class of agents module is responsible for executing agents, managing the event loop, calling the LLM, handling tool calls, and returning structured results.
# In Jupyter, an asyncio event loop is ALREADY running in the background, so we can directly use `await` at the top level of a cell.
response = await Runner.run(
                            starting_agent = Jokester, # The agent instance that should handle the input that defines which model, instructions, tools, and settings are used.
                            input = "Tell a joke about Autonomous AI Agents" # The user input (prompt) sent to the agent.
                            )
print(response.final_output) # The Runner returns a structured result object and the `final_output` is the final text produced by the agent after reasoning, tool calls, and any handoffs are completed.
````

## Example 2: Streaming with Agents
- **Documentation Link: https://openai.github.io/openai-agents-python/streaming/**
```
from openai.agents import Runner

async def main():
    runner = Runner(agent=my_agent)
    async for chunk in runner.run_streamed("Explain quantum computing in simple terms"):
        print(chunk, end="")  # prints each token or chunk as it arrives

import asyncio
asyncio.run(main())
```

````python
from openai.types.responses import ResponseTextDeltaEvent # Imports a specific event class used in streaming responses. ResponseTextDeltaEvent represents: “A partial chunk of text produced by the model”

story_teller = Agent(
            name="StoryTeller",
            instructions="You're a bed-time story teller",
            model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4o-mini",                        
            )

# Runner.run() → waits until the entire response is ready
# Runner.run_streamed() → returns immediately and lets you receive output incrementally and process tokens as they arrive
response = Runner.run_streamed(
                                starting_agent = story_teller,
                                input="Tell me a bed-time story of Ali Baba Chalees chor"
                                )

# Iterating over streamed events
async for event in response.stream_events(): # Opens an asynchronous event stream and each event represents something that happened during execution like partial text output, tool calls, handoffs, final completion signals
    if event.type == "raw_response_event" and isinstance(event.data, ResponseTextDeltaEvent): # Streaming produces many kinds of events, not just text. This condition ensures “Only process events that contain incremental text from the model.”
        print(event.data.delta, end="", flush=True)
````

## Example 4: Getting Structured Output from an Agent
- The default output of an LLM is a free-from text, that works well if you are summarizing an article.
- We have already discussed that the `text` parameter of Responses API can be used to define the format of the model’s textual output.
- Its default is plain `text: {"format": {"type": "text"}}`, which makes the LLM generate free-from text, that works well if you are summarizing an article.
- However, if you are using an LLM into a large software system, where you have multiple components and you want to pass the response of one LLM to another component in a predictable way you may have to use the JSON format. For example, you can set this parameter to `text: {"format": {"type": "json_object"}}` to request the response in structured JSON format, or a user-defined JSON schema
- This is particularly useful when you want to parse and integrate the model’s responses programmatically into applications or data pipelines.
- LLMs often almost get JSON right—but not always. Some common problems are:
    - Extra text like “Here is the JSON you requested”
    - Markdown formatting (```json blocks)
    - Missing fields
    - Incorrect data types (e.g., invalid email format)
- This unpredictability makes raw LLM JSON hard to trust in production. Solution is use `Pydantic` to validate the output generated by LLMs
- **Pydantic** is a Python library for data validation and parsing that uses Python type hints to define data schemas and automatically validates, converts, and enforces types at runtime.

````python
from pydantic import BaseModel
# Define a Pydantic data model named Recipe that inherits from BaseModel, enabling automatic data parsing, type enforcement, and validation for structured user input.
class Recipe(BaseModel):
    title: str                     # Declares a required field `title` that must be a string.
    ingredients: list[str]         # Declares a required field `email` that must be a valid email address.
    cooking_time: int # in minutes
    servings: int

recipe_agent = Agent(
    name="Recipe Agent", 
    instructions=("You are an agent for creating recipes. You will be given the name of a food and your job is to output that as an actual detailed recipe in JSON. The cooking time should be in minutes."),
    model="gpt-4.1-nano",                        
    #model=OpenAIResponsesModel(model="openai/gpt-oss-120b", openai_client= groq_client),    # GPT-oss-20b might not generate  a strictly parsed pydantic model or JSON as requested  and generate an error  
    output_type=Recipe
)

response = await Runner.run(starting_agent=recipe_agent, input="Pakistani Nihari")
rich.print(response.final_output)
````

    - We convert each specialist agent into a tool using the .as_tool() method, specifying:
        - A tool_name that the coordinator will use to reference the tool.
        - A tool_description that helps the coordinator understand when to use this tool.
- This pattern allows you to create complex agent systems while maintaining the separation of concerns. Each specialist agent can have its own set of tools and expertise, while the coordinator manages the user interaction and delegates it to the appropriate specialist.

````python
from agents import Agent, Runner
from dotenv import load_dotenv

load_dotenv()      # loads all the environment variables from a .env file, which is useful for configuration.


# Note Taking Agent Helps users with taking and organizing notes.
note_taking_agent = Agent(
                           name="Note Manager",
                           instructions="You help users take and organize notes efficiently.",
                            model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4.1-nano", 
                           # In a real application, this agent would have note-taking tools
                        )
# Task Management Agent assists users in managing tasks, deadlines, and priorities.
task_management_agent = Agent(
                               name="Task Manager",
                               instructions="You help users manage tasks, deadlines, and priorities.",
                                model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4.1-nano", 
                               # In a real application, this agent would have task management tools
                            )

# Coordinator agent is a higher-level agent that coordinates betwee the above two specialist agents. It uses them as tools to address user requests related to note-taking and task management.
 # The .as_tool() method convert an agent into a tool, specifying: A tool_name that the coordinator will use to reference the tool.A tool_description that helps the coordinator understand when to use this tool.

productivity_assistant = Agent(
                               name="Productivity Assistant",
                               instructions="""You are a productivity assistant that helps users organize their work and personal life.
  
                                               For note-taking questions or requests, use the note_taking tool.
                                               For task and deadline management, use the task_management tool.
  
                                               Help the user decide which tool is appropriate based on their request,
                                               and coordinate between different aspects of productivity.
                                           """,
                               tools=[
                                       note_taking_agent.as_tool(          # The .as_tool() method convert an agent into a tool specifying tool_name and tool_description
                                                                   tool_name="note_taking",          # tool_name is used by coordinator to reference the tool
                                                                   tool_description="For taking, organizing, and retrieving notes and information" # tool_description helps the coordinator understand when to use this tool
                                                               ),
                                       task_management_agent.as_tool(
                                                                       tool_name="task_management",
                                                                       tool_description="For managing tasks, setting deadlines, and tracking priorities"
                                                                   )
                                   ]
                                )
````

### Handoffs vs. agent-as-tool
- The Agents SDK offers two ways for agents to work together: handoffs and using agents as tools:
    - **Use handoffs when:**
        - You want to completely transfer control to another agent
        - The conversation needs to continue with a specialist
        - You’re building a workflow where different agents handle different stages
    - **Use agents-as-tools when:**
        - The primary agent needs to consult a specialist but maintain control
        - You want to incorporate a specialist’s response as part of a larger answer
        - You’re building a hierarchical system where a coordinator delegates subtasks
- Both approaches can be combined in sophisticated systems, with a main agent that sometimes consults specialists (using them as tools) and sometimes hands off control completely when appropriate.
- Handoffs provide a powerful mechanism for building complex agent systems where responsibility transitions between different specialists. When designing your agent architecture, consider which approach best suits your specific use case to create the most effective user experience.

### Version 2: Turning the classifier into an INPUT guardrail

````python
from agents import Agent, Runner
from pydantic import BaseModel

class PoliticalCheck(BaseModel):
    is_political: bool
    reasoning: str

politics_classifier = Agent(
    name="Political Classifier",
    instructions="""
    Check if the user's input contains political opinions,
    elections, political parties, or government criticism.
    """,
    output_type=PoliticalCheck,
    model="gpt-4.1-nano",        # gpt-oss-20b does not work well with strict JSON rules and pydantic
)

# A normal assistant agent (this should NOT answer political questions)
assistant_agent = Agent(
    name="General Assistant",
    instructions="Answer the user's question clearly and concisely.",
    model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4.1-nano", 
)

#query = "what do you think about the fairness of 2018  elections in Pakistan?"
query = "what is the capital of Pakistan?"

# Step 1: Run the guardrail FIRST
guardrail_result = await Runner.run(
    starting_agent=politics_classifier,
    input=query
)

# Step 2: Enforce the input guardrail
if guardrail_result.final_output.is_political:
    print("🚫 Input blocked by guardrail")
    print("Reason:", guardrail_result.final_output.reasoning)
else:
    # Step 3: Only now send input to the main model
    response = await Runner.run(
        starting_agent=assistant_agent,
        input=query
    )
    print(response.final_output)
````

### Version 3: Clean production-style guardrail pipeline

````python
from agents import Agent, Runner
from pydantic import BaseModel

# ----------------------------
# Guardrail output definition
# ----------------------------
class PoliticalCheck(BaseModel):
    is_political: bool
    reasoning: str

# ----------------------------
# Guardrail agent
# ----------------------------
politics_guardrail_agent = Agent(
    name="Politics Input Guardrail",
    instructions="Identify whether the user input contains political content such as elections, political parties, governance, or public policy.",
    output_type=PoliticalCheck,
    model="gpt-4.1-nano"
)

# ----------------------------
# Main assistant agent
# ----------------------------
assistant_agent = Agent(
    name="Safe Assistant",
    instructions="""
    Answer the user's question helpfully.
    Do not answer political questions.
    """,
    model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4.1-nano", 
)

# ----------------------------
# Guardrail pipeline function
# ----------------------------
async def run_with_input_guardrail(user_input: str):
    # Step 1: Check input safety
    guardrail = await Runner.run(
        starting_agent=politics_guardrail_agent,
        input=user_input
    )

    if guardrail.final_output.is_political:
        return {
            "blocked": True,
            "reason": guardrail.final_output.reasoning
        }

    # Step 2: Safe to proceed
    response = await Runner.run(
        starting_agent=assistant_agent,
        input=user_input
    )

    return {
        "blocked": False,
        "response": response.final_output
    }

# ----------------------------
# Try it out
# ----------------------------
# query = "What is the capital of Pakistan?"
query = "what do you think about the fairness of 2018  elections in Pakistan?"
result = await run_with_input_guardrail(query)

if result["blocked"]:
    print("🚫 Blocked:", result["reason"])
else:
    print("✅ Response:", result["response"])
````

## Example 11: Output Guardrails
OpenAI have also included guardrails in the Agents SDK. These come as input guardrails and output guardrails, the input_guardrail checks that the input going into your LLM is "safe" and the output_guardrail checks that the output from your LLM is "safe".

Let's see how to use them. First, we'll implement a guardrail powered by another LLM (more tokens means more $$$ for OpenAI).


````python
answering_agent = Agent(
    name="Answering Agent",
    instructions="Answer the user's question clearly.",
    model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4.1-nano", 
)
````

````python
class OutputCheck(BaseModel):
    is_allowed: bool
    reason: str

output_guardrail_agent = Agent(
    name="Output Guardrail",
    #instructions="Check whether the assistant's response contains political content such as elections, political parties, governance, or public policy.",
    instructions="""
    Determine whether the assistant's response contains ANY political content,
    including:
    - Elections
    - Political parties
    - Government actions
    - Political opinions (even neutral or balanced)

    If political content is present, set is_allowed = false.
    """,
    output_type=OutputCheck,
    model=OpenAIResponsesModel(model="openai/gpt-oss-20b", openai_client= groq_client),       # model="gpt-4.1-nano", 
)
````

````python
async def run_with_output_guardrail(user_input: str):
    # Step 1: Generate response
    answer_result = await Runner.run(
        starting_agent=answering_agent,
        input=user_input
    )

    generated_text = answer_result.final_output

    # Step 2: Validate output
    guardrail_result = await Runner.run(
        starting_agent=output_guardrail_agent,
        input=generated_text
    )

    if not guardrail_result.final_output.is_allowed:
        return {
            "blocked": True,
            "reason": guardrail_result.final_output.reason
        }

    return {
        "blocked": False,
        "response": generated_text
    }
````

````python
# query = "What is the capital of Pakistan?"
query = "what do you think about the fairness of 2018  elections in Pakistan?"
result = await run_with_input_guardrail(query)


result = await run_with_output_guardrail(query)

if result["blocked"]:
    print("🚫 Output blocked:", result["reason"])
else:
    print("✅ Final response:", result["response"])
````

## Phase 2: Converting Agents to Tools

Now comes the interesting part! We're going to **convert our agents into tools** that other agents can use.

### What Does "Agent as Tool" Mean?

Think of it this way:

**Before (Phase 1):** We had 3 standalone agents. To use them, we'd run `Runner.run(agent, prompt)` for each one.

**Now (Phase 2):** We're converting these agents into **tools** that another agent can call automatically.

It's like the difference between:
- **Calling a friend directly** (Phase 1: You run the agent yourself)
- **Giving someone your friend's phone number** (Phase 2: Another agent can call this agent as a tool)

### Understanding `.as_tool()`

The `.as_tool()` method transforms an agent into a callable tool. Here's what it does:
```python
tool = agent.as_tool(
    tool_name="unique_tool_name",           # How other agents will refer to this tool
    tool_description="When to use this"     # Helps other agents decide when to call it
)
```

**What happens under the hood?**

1. **Wraps the agent** in a function that other agents can call
2. **Creates a tool signature** with name and description
3. **Makes it callable** by the LLM through function calling
4. **Returns the agent's output** when invoked

**Why is `tool_description` important?**

The description helps the manager agent (which we'll build next) decide **when** to use this tool. It's like writing instructions:

- Bad: `"Writes emails"` (Too vague - when would you use this?)
- Good: `"Write formal emails to professors for meeting requests and grade inquiries"` (Clear when to use!)

````python
# Convert each email agent into a tool

# Tool descriptions that explain when to use each agent
description1 = "Write a formal email to a professor (for meeting requests, clarifications, recommendation letters, grade inquiries)"

description2 = "Write a friendly email to classmates or peers (for group projects, study sessions, homework collaboration)"

description3 = "Write a professional email to administration staff (for registration, deadlines, documentation requests)"

# Convert agents to tools using .as_tool()
professor_tool = professor_email_agent.as_tool(
    tool_name="professor_email_tool", 
    tool_description=description1
)

peer_tool = peer_email_agent.as_tool(
    tool_name="peer_email_tool", 
    tool_description=description2
)

admin_tool = admin_email_agent.as_tool(
    tool_name="admin_email_tool", 
    tool_description=description3
)

# Collect all tools in a list
email_writing_tools = [professor_tool, peer_tool, admin_tool]
````

````python
# ==================== Tool 3: Email Sender (Real Email with SendGrid) ====================

@function_tool  # NEW: Decorator that converts a Python function into a tool
def send_email(subject: str, formatted_body: str, recipient_email: str) -> Dict[str, str]:  # NEW: Type hints for parameters and added recipient_email
    """
    Sends an actual email using SendGrid API.
    
    Args:
        subject: The email subject line
        formatted_body: The formatted email body with greeting, content, and signature
        recipient_email: The email address to send to
    
    Returns:
        A dictionary with status and message
    """
    try:
        # Initialize SendGrid client with API key from environment variable
        sg = sendgrid.SendGridAPIClient(api_key=os.environ.get('SEND_GRID'))  # NEW: Create SendGrid client
        
        # Configure email sender (REPLACE WITH YOUR VERIFIED SENDER EMAIL!)
        from_email = Email("zalaidbutt1@gmail.com")  # TODO: Change to your verified sender email
        
        # Configure email recipient
        to_email = To(recipient_email)  # NEW: Use the recipient email passed as parameter
        
        # Create email content (plain text version)
        content = Content("text/plain", formatted_body)  # NEW: Create email content
        
        # Build the email
        mail = Mail(from_email, to_email, subject, content)  # NEW: Construct the email object
        
        # Send the email
        response = sg.client.mail.send.post(request_body=mail.get())  # NEW: Send via SendGrid API
        
        # Print confirmation to console
        print("=" * 50)
        print("EMAIL SENT VIA SENDGRID!")
        print("=" * 50)
        print(f"To: {recipient_email}")
        print(f"Subject: {subject}")
        print(f"Status Code: {response.status_code}")  # NEW: Show HTTP status code
        print("=" * 50)
        
        return {  # NEW: Return structured response
            "status": "success", 
            "message": f"Email sent successfully to {recipient_email}!",
            "status_code": response.status_code
        }
    
    except Exception as e:  # NEW: Error handling for SendGrid failures
        # If something goes wrong, return error details
        print(f" Error sending email: {str(e)}")
        return {
            "status": "error",
            "message": f"Failed to send email: {str(e)}"
        }
````

## Phase 3: Building the Manager Agent (The Orchestrator)

Now we'll create the **"brain"** of our system - the **Student Communication Manager Agent**.

### What Does This Agent Do?

This is the **only agent the student interacts with**. It:

1. **Analyzes** the student's request to understand who the email is for
2. **Selects** the right email writing tool (professor, peer, or admin)
3. **Generates** the email draft using that tool
4. **Hands off** to the Email Composer Agent for finalization and sending

### The Power of Handoffs

The `handoffs` parameter allows this agent to **transfer control** to another agent:
```python
handoffs = [email_composer_agent]
```

When the manager says "I'm done, here's the email," it hands off to the Email Composer Agent, which then takes over to finalize and send.
