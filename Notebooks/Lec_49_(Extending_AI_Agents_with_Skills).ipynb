# <span style='background :lightgreen' >1. Recap: Where Skills Fit in an Agentic System</span>

> **To be written.** Planned content:
- Back to the key-components figure from **Lec-39, Section 3** (`agent-new-06.png`): skills sit in the green **Context** group, next to instructions and memory.
- **The problem:** putting every procedure (report formats, coding conventions, step-by-step workflows) into the system prompt makes it long and costly, because it is resent on every call (the context-window growth seen in Lec-39, Example 2c), and the model is distracted by instructions it does not need.
- **The idea of a skill:** package each procedure separately, and let the agent load it **only when a task needs it**.

# <span style='background :lightgreen' >2. What is an Agent Skill?</span>

> **To be written.** Planned content:
- **Definition:** a skill is a **folder** containing a `SKILL.md` file, optionally with `scripts/`, `references/` and `assets/` sub-folders. It teaches an agent *how* to do one kind of task well, like a standard operating procedure (SOP).
- **Anatomy of `SKILL.md`:** YAML front matter with a `name` and a `description`, followed by Markdown instructions. Show a complete small example.
- **The open standard:** introduced by Anthropic for Claude in October 2025 and published as an open standard at **agentskills.io** in December 2025; now supported by Claude, OpenAI Codex, GitHub Copilot, Gemini CLI, Cursor and other agents. Verify the current list of supporting agents when writing.

# <span style='background :lightgreen' >3. Progressive Disclosure: How an Agent Loads a Skill</span>

> **To be written.** Planned content:
- **New figure** (in the `agent-new-*` style) showing the three levels:
    - **Level 1, at start-up:** only each skill's `name` and `description` go into the context (a few dozen tokens per skill).
    - **Level 2, when relevant:** the agent reads the full `SKILL.md`.
    - **Level 3, only if needed:** the agent opens reference files or runs bundled scripts.
- **Why it matters:** an agent can have dozens of skills installed while its context stays small. Link to context engineering and context rot (Lec-39).

# <span style='background :lightgreen' >8. Skills vs. Tools vs. MCP vs. System Prompts vs. Sub-Agents</span>

> **To be written.** Planned content:
- A comparison table: what each one is, when it is loaded, who executes it, and a typical example.
- **Rules of thumb:** a **tool** is a single action; an **MCP server** delivers tools (and data) from another program; a **skill** is know-how that tells the agent how to combine them; the **system prompt** holds what applies to *every* task; a **sub-agent** (Lec-48) gets its own context for a big sub-task.

# <span style='background :lightgreen' >9. Security: Skills Are Instructions and Code</span>

> **To be written.** Planned content:
- A skill can contain **instructions** (which can carry prompt injections) and **scripts** (which run with your permissions).
- Only install skills from sources you trust, read `SKILL.md` and the scripts before installing, and run skill scripts in a sandbox or with approval.
